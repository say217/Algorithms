# What a list really is

A list is a sequence of references stored contiguously in memory.
It does not store values directly; it stores pointers to objects.
That’s why a list can mix types and why mutation behaves the way it does.

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)   # [1, 2, 3, 4]
````

````python
nums = [1, 2, 3]
mixed = [1, "hi", 3.14, True]
empty = []
via_constructor = list("abc")   # ['a', 'b', 'c']
via_constructor
````

````python
#Indexing (positive & negative)
a = [10, 20, 30, 40]

a[0]    # 10
a[2]    # 30
a[-1]   # 40 (last element)
a[-2]   # 30
````

````python
#Slicing (the most misunderstood power tool)
a = [0, 1, 2, 3, 4, 5]

a[1:4]     # [1, 2, 3]
a[:3]      # [0, 1, 2]
a[3:]      # [3, 4, 5]
a[::2]     # [0, 2, 4]
a[::-1]    # reverse
````

````python
#Mutability (why lists are dangerous and powerful)
a = [1, 2, 3]
a[0] = 100
print(a)   # [100, 2, 3]
````

````python
a = [1, 2]


a.append(3)# O(1) amortized
print(a)
a.extend([4, 5, 4, 5, 2])   # adds multiple
print(a)
a.insert(0, 99)    # O(n) (shifts elements)
print(a)
````

### Why list comprehension is safe
```
a = [x for x in a if x % 2 != 0]
```

What actually happens:

- Python iterates over the original list
- Builds a brand new list in memory
- Assigns that new list back to a

````python
a = [1, 2, 3, 4]
for x in a:
    if x % 2 == 0:
        a.remove(x)
print(a)
````

````python
a = [x for x in a if x % 2 != 0]
a
````

### Default mutable argument (killer interview bug)


```
lst points to the stored default list []
Append 1
List becomes [1]
print(add_item(2))

No new list is created
lst points to the same list as before
Append 2


def add_item(x, lst=[]):
    lst.append(x)
    return lst

print(add_item(1))
print(add_item(2))



```
Here’s the key difference:
None is immutable
Safe to share
Signals “no value provided”
Eachcall where lst is omitted:
lst starts as None
You explicitly create a new list
That list lives only for this call

````python
def add_item(x, lst=None):
    if lst is None:
        lst = []
    lst.append(x)
    return lst

print(add_item(1))
print(add_item(2))
    
````

### Clearing list references

a = []
Python does not touch the old list.

Instead:

It creates a new empty list

Rebinds the name a to it

Now memory looks like:

[1, 2, 3]  ←── b

[]         ←── a

````python

a = [1, 2, 3]
b = a
a = []
print(b)
````
