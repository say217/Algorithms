# 📚 Data Structures

## What is a Data Structure?

A **data structure** is a way of organizing, managing, and storing data so that it can be accessed and modified efficiently. It defines the relationship between the data and the operations that can be performed on it.

---

## 🧩 Types of Data Structures

### 1. **Linear Data Structures**
These store data in a sequential manner.

- **Array**: A collection of elements identified by index.
- **Linked List**: A series of connected nodes (each with data and a pointer).
- **Stack**: Follows LIFO (Last In, First Out) principle.
- **Queue**: Follows FIFO (First In, First Out) principle.

---

### 2. **Non-Linear Data Structures**
These don’t store data sequentially.

- **Tree**: Hierarchical structure with nodes (e.g., Binary Tree, Binary Search Tree).
- **Graph**: A set of nodes connected by edges (can be directed or undirected).

---

### 3. **Hash-Based Data Structures**
These use a hash function to map keys to values.

- **Hash Table / Hash Map**: Stores key-value pairs for fast retrieval.

---

### 4. **Advanced Data Structures**
Used for specialized operations and optimized performance.

- **Heap**: A complete binary tree used in priority queues.
- **Trie (Prefix Tree)**: Efficient for word searches in dictionaries.
- **Segment Tree / Fenwick Tree (Binary Indexed Tree)**: Used for range queries.

---

## 🔁 Summary Table

|--------------|----------------------------------------|------------------------------------|

| Non-Linear   | Tree, Graph                            | Hierarchical or networked          |
| Hash-Based   | Hash Table, Hash Map                   | Fast lookup with keys              |
| Advanced     | Heap, Trie, Segment Tree, Fenwick Tree | Specialized and efficient access   |

# 📦 Array

## What is an Array?

An **Array** is a **linear data structure** that stores elements of the **same data type** in **contiguous memory locations**. Each element can be accessed directly using its **index**.

---

## 🧠 Key Characteristics

- **Fixed Size**: The size of an array is defined at the time of declaration.
- **Indexed Access**: Each element is accessed using its index (starting from 0).
- **Homogeneous Data**: All elements are of the same data type.
- **Stored in Contiguous Memory**: Enhances performance when iterating.

---

## 📌 Syntax Examples

### 🔹 In C:
```c

```

### 🔹 In Python:
```python

```

### 🔹 In Java:

## 🎯 Basic Operations

| Operation         | Description                        | Time Complexity |
|------------------|------------------------------------|-----------------|
| Access           | Retrieve an element by index       | O(1)            |
| Insertion        | Insert at a position (may shift)   | O(n)            |
| Deletion         | Delete from a position             | O(n)            |
| Search           | Find an element                    | O(n)            |
| Traversal        | Visit all elements                 | O(n)            |

---

## ✅ Advantages

- Fast access to elements using index.
- Easy to implement and use.
- Memory locality improves cache performance.

---

## ❌ Disadvantages

- Fixed size: Cannot grow or shrink dynamically.
- Insertion and deletion are costly (need shifting).
- Wastes memory if size is overestimated.

# Selection sort

````python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j
                
        arr[i], arr[min_index] = arr[min_index], arr[i]


arr = [12, 45, 22, 43, 10, 90]
selection_sort(arr)
print('sorted array', arr)
        
            
````

# Incertion sort


````python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j= i - 1

        while j >=0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j = j - 1

        arr[j + 1] = key  
# Example
arr = [12, 11, 13, 5, 6]
insertion_sort(arr)
print("Sorted array:", arr)        
````

# Insertion Sort in Python

This document explains the working of the `insertion_sort` function implemented in Python.

### Purpose
The insertion_sort function sorts an array arr in ascending order using the insertion sort algorithm.
# How It Works

### Outer Loop (`for i in range(1, len(arr))`):

- Iterates over each element starting from index 1 (since a single-element list is already "sorted").
- `key = arr[i]`: The current element to be inserted into the sorted portion of the array.
- `j = i - 1`: Initialize the pointer to the last index of the sorted portion.

### Inner Loop (`while j >= 0 and arr[j] > key`):

- Checks if elements in the sorted portion are greater than the key.
- If so, it shifts the element at `arr[j]` one position to the right (`arr[j + 1] = arr[j]`).
- Decrements `j` to continue checking previous elements.

### Insertion (`arr[j + 1] = key`):

- Once the correct position is found (i.e., where `arr[j] <= key`), the key is placed at `arr[j + 1]`.

---

# Example

Given the list:

```csharp
[12, 11, 13, 5, 6]
```
#### Sorting Steps:
- i = 1: key = 11 → insert before 12 → [11, 12, 13, 5, 6]

- i = 2: key = 13 → already in correct position → [11, 12, 13, 5, 6]

- i = 3: key = 5 → insert before 11 → [5, 11, 12, 13, 6]

- i = 4: key = 6 → insert between 5 and 11 → [5, 6, 11, 12, 13]


# Merge sort


````python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left)
        merge_sort(right)

        i = k = j = 0

        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i = i + 1
            else:
                arr[k] = right[j]
                j = j + 1
            k = k + 1

            while i < len(left):
                arr[k] = left[i]
                i = i + 1
                k = k + 1

            while j < len(right):
              arr[k] = right[j]
              j += 1
              k += 1

arr = [9, 24, 12, 89, 43, 90, 8]
merge_sort(arr)

print('sorted array', arr)


                
````

# Merge Sort: Key Idea

Merge Sort uses a **Divide and Conquer** strategy:

- **Divide** the array into halves recursively until each subarray has one element.
- **Conquer** by merging the smaller sorted arrays into larger sorted arrays.
- Continue until you get a fully sorted array.

---

## 🔸 Step-by-Step Explanation

### 🧩 Step 1: Split the Array

We start by splitting the array:

**Original:**  
`[38, 24, 12, 89, 43, 90]`

**Split into:**

- Left: `[38, 24, 12]`
- Right: `[89, 43, 90]`

---

### 🧩 Step 2: Recurse on Left Half `[38, 24, 12]`

Split again:

- `[38]` (base case)
- `[24, 12]` → split into `[24]` and `[12]`

**Now merge `[24]` and `[12]`:**

- Compare 24 and 12 → 12 is smaller → `[12, 24]`

**Now merge `[38]` and `[12, 24]`:**

- Compare 38 and 12 → 12  
- Compare 38 and 24 → 24  
- Remaining 38 → `[12, 24, 38]`

---

### 🧩 Step 3: Recurse on Right Half `[89, 43, 90]`

Split again:

- `[89]` (base case)
- `[43, 90]` → split into `[43]` and `[90]`

**Merge `[43]` and `[90]`:**

- 43 < 90 → `[43, 90]`

**Now merge `[89]` and `[43, 90]`:**

- 43 < 89 → 43  
- 89 < 90 → 89  
- Remaining 90 → `[43, 89, 90]`

---

### 🧩 Final Merge: `[12, 24, 38]` and `[43, 89, 90]`

- Compare 12 and 43 → 12  
- Compare 24 and 43 → 24  
- Compare 38 and 43 → 38  
- Remaining 43, 89, 90 → `[12, 24, 38, 43, 89, 90]`

---

# Quick Sort


````python
def partition(arr, low, high):
 
    '''
    Choose the last element as pivot, then
    rearrange arr[low..high] so that:
    - elements ≤ pivot come before it
    - elements > pivot come after it
    Returns the final pivot index.
   '''
    pivot = arr[high]
    i = low
    for j in range(low, high):
        if arr[j] <= pivot:
            # swap arr[i] and arr[j]
            arr[i], arr[j] = arr[j], arr[i]
            i = i + 1
    arr[i], arr[high] = arr[high], arr[i]
    return i

def sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)
        sort(arr, low, pi - 1)
        sort(arr, pi + 1, high)

arr = [22, 12, 34, 54, 11]
sort(arr, 0, len(arr) - 1)
print('sorted array', arr)
````

# Seconed largest and smallest

````python
def second_largest(arr):
    if len(arr) < 2:
        return None  # Not enough elements

    largest = second = float('-inf')

    for num in arr:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num != largest:
            second = num

    return second if second != float('-inf') else None

# Example usage:
arr = [5, 1, 8, 3, 8, 2]
print("Second largest element is:", second_largest(arr))
````

# Find Subarray With given sum

````python
def find_subarray(arr, target):
    n = len(arr)
    for start in range(n):
        current_sum = 0
        for end in range(start, n):
            current_sum += arr[end]

            if current_sum == target:
              return arr[start:end+1]
            if current_sum > target:
                break
    return None

arr = [12, 44, 45, 1, 3, 4, 5]
d = 13
print('sub array', find_subarray(arr, d))
````

````python
def find_all_subarrays_with_sum(arr, target):
    n = len(arr)
    result = []

    for start in range(n):
        current_sum = 0
        for end in range(start, n):
            current_sum += arr[end]

            if current_sum == target:
                # Append the subarray (slice from start to end)
                result.append(arr[start:end+1])

            # Optional optimization for positive numbers
            if current_sum > target:
                break

    return result

# Example usage
arr = [1, 2, 3, 4, 5]
target = 5

all_subarrays = find_all_subarrays_with_sum(arr, target)

if all_subarrays:
    print(f"All subarrays with sum {target}:")
    for sub in all_subarrays:
        print(sub)
else:
    print("No subarray found")
````

````python
def find_combinations(arr, target):
    result = []

    def backtrack(start, path, current_sum):
        if current_sum == target:
            result.append(path[:])
            return
        if current_sum > target:
            return
        
        for i in range(start, len(arr)):
            path.append(arr[i])
            backtrack(i + 1, path, current_sum + arr[i])
            path.pop()

    backtrack(0, [], 0)
    return result

arr = [1, 2, 3, 4, 5]
target = 5
combinations = find_combinations(arr, target)

for combo in combinations:
    print(combo)
````

# Kadens Algorith find max subarray sum
🧠 Problem:
Given an integer array nums, find the contiguous subarray with the largest sum and return its sum.
```
Input: [-2,1,-3,4,-1,2,1,-5,4]  
Output: 6  # Subarray: [4,-1,2,1]
```

````python
def max_subarray_sum(nums):
    max_sum = nums[0]
    current_sum = nums[0]

    for num in nums[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)

    return max_sum

# Example
nums = [5, -3, 4, -1, 2, -6, 1]
print("Max Subarray Sum:", max_subarray_sum(nums))
````

````python
def max_subarray_sum(nums):
    max_sum = current_sum = nums[0]
    start = end = temp_start = 0

    for i in range(1, len(nums)):
        if nums[i] > current_sum + nums[i]:
            current_sum = nums[i]
            temp_start = i
        else:
            current_sum += nums[i]

        if current_sum > max_sum:
            max_sum = current_sum
            start = temp_start
            end = i

    return max_sum, nums[start:end+1], (start, end)

# Example
nums = [5, -3, 4, -1, 2, -6, 1]
max_sum, subarray, indices = max_subarray_sum(nums)

print("Max Subarray Sum:", max_sum)
print("Subarray:", subarray)
print("Indices:", indices)
````

#  Product of Array Except Self
🧠 Problem:
Given an integer array nums, return an array output such that output[i] is equal to the product of all the elements of nums except nums[i], without using division and in O(n) time.
```
Input: [1, 2, 3, 4]  
Output: [24, 12, 8, 6]
```

````python
def product_except_self(nums):
    n = len(nums)
    
    # Initialize result arrays
    left = [1] * n     # Left product
    right = [1] * n    # Right product
    output = [1] * n   # Final output

    # Build the left product array
    for i in range(1, n):
        left[i] = left[i - 1] * nums[i - 1]

    # Build the right product array
    for i in range(n - 2, -1, -1):
        right[i] = right[i + 1] * nums[i + 1]

    # Multiply left and right products
    for i in range(n):
        output[i] = left[i] * right[i]

    return output

# Example
nums = [1, 2, 3, 4]
print("Output:", product_except_self(nums))
````

````python
def move_zeros(nums):
    pos = 0  # next position for non-zero

    for num in nums:
        if num != 0:
            nums[pos] = num
            pos += 1

    while pos < len(nums):
        nums[pos] = 0
        pos += 1

    return nums

# Example
print(move_zeros([0, 1, 0, 3, 12]))
````

````python
def move_zeros_to_front(nums):
    pos = len(nums) - 1  # next position for non-zero from the right

    # Place non-zero elements from the right side
    for num in reversed(nums):
        if num != 0:
            nums[pos] = num
            pos -= 1

    # Fill the remaining front positions with zeros
    while pos >= 0:
        nums[pos] = 0
        pos -= 1

    return nums

# Example
print(move_zeros_to_front([0, 1, 0, 3, 12]))
````

# . Merge Intervals
Given a collection of intervals, merge all overlapping intervals.
```
Example:
Input: [[1,3],[2,6],[8,10],[15,18]]
Output: [[1,6],[8,10],[15,18]]
```

````python
def merge_intervals(intervals):
    if not intervals:
        return []

    intervals.sort()  # Sort by first element by default
    merged = []

    for interval in intervals:
        if not merged or merged[-1][1] < interval[0]:
            merged.append(interval)
        else:
            merged[-1][1] = max(merged[-1][1], interval[1])

    return merged

# Example usage
input_intervals = [[1, 3], [2, 6], [8, 10], [15, 18]]
print(merge_intervals(input_intervals))  # Output: [[1, 6], [8, 10], [15, 18]]
````

#  Longest Consecutive Sequence
Find the length of the longest consecutive elements sequence.

Example:
Input: [100, 4, 200, 1, 3, 2] → Output: 4 (sequence is 1,2,3,4)
✅ Asked in: Google, Facebook

````python
def longestConsecutive(nums):
    nums_set = set(nums)
    longest = 0

    for num in nums_set:
        # Start of a sequence
        if num - 1 not in nums_set:
            current_num = num
            count = 1

            while current_num + 1 in nums_set:
                current_num += 1
                count += 1

            if count > longest:
                longest = count

    return longest
print(longestConsecutive([100, 4, 200, 1, 3, 2]))
# Output: 4 (The sequence is 1, 2, 3, 4)

````

# Rearrange
```
Input: arr[] = [1, 2, 3, 4, 5, 6]
Output: [6, 1, 5, 2, 4, 3]
Explanation: Max element = 6, min = 1, second max = 5, second min = 2, and so on... The modified array is: [6, 1, 5, 2, 4, 3]
```

````python
def rearrange(arr):
    n = len(arr)
    max_elem = max(arr) + 1  # max_elem > any element in arr
    
    max_idx = n - 1  # pointer to max element
    min_idx = 0      # pointer to min element
    
    for i in range(n):
        if i % 2 == 0:
            # place max element at even index
            arr[i] += (arr[max_idx] % max_elem) * max_elem
            max_idx -= 1
        else:
            # place min element at odd index
            arr[i] += (arr[min_idx] % max_elem) * max_elem
            min_idx += 1
    
    # decode the new values
    for i in range(n):
        arr[i] = arr[i] // max_elem

# Example usage:
arr = [1, 2, 3, 4, 5, 6]
rearrange(arr) 
print(arr)  # Output: [6, 1, 5, 2, 4, 3]
````

Note : Inversion count is the number of pairs of elements (i, j) such that i < j and arr[i] > arr[j].

Examples:
```
Input: arr[] = [2, 4, 1, 3, 5]

Input: arr[] = [2, 3, 4, 5, 6]

```

````python
def find_inversion_pairs(arr):
    n = len(arr)
    inversions = []
    count = 0
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] > arr[j]:
                inversions.append((i, j))
                count += 1
    return inversions, count

# Example usage
arr = [2, 4, 1, 3, 5]
inversion_pairs = find_inversion_pairs(arr)
print("Inversion pairs:", inversion_pairs)
````

# Find leaders of an array

````python
def find_leaders(arr):
    n = len(arr)
    leaders = []
    max_from_right = arr[-1]
    leaders.append(max_from_right)

    for i in range(n - 2, -1, -1):
        if arr[i] > max_from_right:
            leaders.append(arr[i])
            max_from_right = arr[i]

    return leaders[::-1]  # Reverse to maintain original order
arr = [16, 17, 4, 3, 5, 2]
print(find_leaders(arr))  # Output: [17, 5, 2]

````

# Reverse an array in group

````python
def reverse_in_groups(arr, k):
    n = len(arr)
    for i in range(0, n, k):
        left = i
        right = min(i + k - 1, n - 1)

        # Reverse sub-array from left to right
        while left < right:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1
            
arr = [1, 2, 3, 4, 5]
k = 4
reverse_in_groups(arr, k)
print(arr)  # Output: [3, 2, 1, 5, 4]

````

# Pythogorian
Given an array arr[], return true if there is a triplet (a, b, c) from the array (where a, b, and c are on different indexes) that satisfies a2 + b2 = c2, otherwise return false.

Examples:
```
Input: arr[] = [3, 2, 4, 6, 5]
Output: true
Explanation: a=3, b=4, and c=5 forms a pythagorean triplet.

````python
def has_triplet(arr):
    n = len(arr)
    # Square all elements
    arr = [x*x for x in arr]
    # Sort the squared elements
    arr.sort()

    # Try each element as c^2 (largest), check if a^2 + b^2 = c^2
    for i in range(n-1, 1, -1):
        a = 0
        b = i - 1
        while a < b:
            if arr[a] + arr[b] == arr[i]:
                return True
            elif arr[a] + arr[b] < arr[i]:
                a += 1
            else:
                b -= 1
    return False

# Example
arr = [3, 2, 4, 6, 5]
print(has_triplet(arr))  # Output: True
````

# Array Manipulation (Hard)
hackerrank.comstackoverflow.comstackoverflow.com
Description: Start with a 1-indexed array of zeros. Perform m operations where each operation adds a value k to all elements between indices a and b (inclusive). Return the maximum value in the array after all operations. This problem often requires optimization using prefix sum techniques to avoid timeouts.
Example Input: Array size n=5, queries [[1,2,100], [2,5,100], [3,4,100]]
Example Output: 200 (maximum value after operations)
```
Apply [1, 2, 100] → add 100 to indices 1 and 2
→ arr = [100, 100, 0, 0, 0]

Apply [2, 5, 100] → add 100 to indices 2, 3, 4, 5
→ arr = [100, 200, 100, 100, 100]

Apply [3, 4, 100] → add 100 to indices 3, 4
→ arr = [100, 200, 200, 200, 100]

✅ Max value = 200

````python
def arrayManipulation(n, queries):
    arr = [0] * (n + 2)  # We use n+2 to handle b+1 without index error

    for a, b, k in queries:
        arr[a] += k
        arr[b + 1] -= k

    max_val = 0
    current = 0
    for i in range(1, n + 1):
        current += arr[i]
        if current > max_val:
            max_val = current

    return max_val

# Example usage:
n = 5
queries = [[1, 2, 100], [2, 5, 100], [3, 4, 100]]
print(arrayManipulation(n, queries))  # Output: 200
````

# Sparse Arrays (Medium)stackoverflow.com
Description: Given a list of input strings and a list of query strings, determine how many times each query string occurs in the input strings. Return an array of the frequencies.
Example Input: Strings = ['ab', 'ab', 'abc'], Queries = ['ab', 'abc', 'bc']
Example Output: [2, 1, 0]


````python
def matchingStrings(strings, queries):
    # Step 1: Count frequencies
    freq = {}
    for s in strings:
        if s in freq:
            freq[s] += 1
        else:
            freq[s] = 1

    # Step 2: Get results directly using dictionary lookup
    result = []
    for q in queries:
        if q in freq:
            result.append(freq[q])
        else:
            result.append(0)

    return result

strings = ['ab', 'ab', 'abc']
queries = ['ab', 'abc', 'bc']
print(matchingStrings(strings, queries))  # Output: [2, 1, 0]
````

# Longest Consecutive Sequence

````python
def longest_consecutive(nums):
    if not nums:
        return 0

    nums = list(set(nums))  # Remove duplicates
    nums.sort()             # Sort the list
    max_count = 1
    count = 1

    for i in range(1, len(nums)):
        if nums[i] == nums[i - 1] + 1:
            count += 1
        else:
            count = 1
        if count > max_count:
            max_count = count

    return max_count
nums = [100, 4, 200, 1, 3, 2]
print(longest_consecutive(nums))  # Output: 4
````

# 18. Merge Intervals (LeetCode #56)


````python
def merge_intervals(intervals):
    if not intervals:
        return []

    # Step 1: Sort intervals by starting number
    intervals.sort()
    merged = [intervals[0]]

    # Step 2: Go through the list and merge when needed
    for current in intervals[1:]:
        last = merged[-1]
        # If intervals overlap, merge them
        if current[0] <= last[1]:
            last[1] = max(last[1], current[1])
        else:
            merged.append(current)

    return merged
intervals = [[1,3], [2,6], [8,10], [15,18]]
print(merge_intervals(intervals))  # Output: [[1,6], [8,10], [15,18]]

````

## Subarray Sum Equals K

````python
def subarray_sum(nums, k):
    count = 0

    # Go through all possible subarrays
    for start in range(len(nums)):
        total = 0
        for end in range(start, len(nums)):
            total += nums[end]
            if total == k:
                count += 1

    return count

nums = [1, 2, 3]
k = 3
print(subarray_sum(nums, k))  # Output: 2
````

# Maximum Product Subarray

````python
def max_product(nums):
    if not nums:
        return 0

    # Start with the first number
    max_so_far = nums[0]
    current_max = nums[0]
    current_min = nums[0]

    for num in nums[1:]:
        # If num is negative, swap max and min
        if num < 0:
            current_max, current_min = current_min, current_max

        # Calculate current max and min
        current_max = max(num, current_max * num)
        current_min = min(num, current_min * num)

        # Update result
        max_so_far = max(max_so_far, current_max)

    return max_so_far

nums = [2, 3, -2, 4]
print(max_product(nums))  # Output: 6
````
