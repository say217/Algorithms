```python
#Merge Two Sorted Lists (Without sorted())
def merge_sorted_lists(a, b):
    i = j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    result.extend(a[i:])
    result.extend(b[j:])
    return result


list1 = ["apple", "grape", "pear"]
list2 = ["banana", "orange", "peach"]

print(merge_sorted_lists(list1, list2))
```

```python
# Merge Two Lists and Remove Duplicates (Keep Order)



def merge_unique(list1, list2):
    seen = set()
    result = []

    for item in list1 + list2:
        if item not in seen:
            seen.add(item)
            result.append(item)

    return result


a = ["a", "b", "c", "a"]
b = ["b", "d", "e", "a"]

print(merge_unique(a, b))
```

```python
# . Merge Two Strings Alternately



def merge_strings(s1, s2):
    result = []

    n = max(len(s1), len(s2))

    for i in range(n):
        if i < len(s1):
            result.append(s1[i])
        if i < len(s2):
            result.append(s2[i])

    return "".join(result)


print(merge_strings("abcdef", "123"))
```

```python
# . Merge K Sorted Lists

import heapq

def merge_k_lists(lists):
    return list(heapq.merge(*lists))


lists = [
    ["apple", "grape"],
    ["banana", "pear"],
    ["cherry", "orange"]
]

print(merge_k_lists(lists))
```

```python
# Merge Intervals



def merge_intervals(intervals):
    intervals.sort()

    merged = [intervals[0]]

    for start, end in intervals[1:]:
        last_end = merged[-1][1]

        if start <= last_end:
            merged[-1][1] = max(last_end, end)
        else:
            merged.append([start, end])

    return merged


intervals = [[1,3], [2,6], [8,10], [9,12]]

print(merge_intervals(intervals))
```
