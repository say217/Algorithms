````python
def quick_sort(chars):
    if len(chars) <= 1:
        return chars

    pivot = chars[0]
    left = []
    right = []

    for ch in chars[1:]:
        if ch <= pivot:
            left.append(ch)
        else:
            right.append(ch)

    return quick_sort(left) + [pivot] + quick_sort(right)
s = "banana"

chars = list(s)
sorted_chars = quick_sort(chars)
sorted_string = "".join(sorted_chars)

print(sorted_string)
````

````python
def are_anagrams(s1, s2):
    # Step 1: length check
    if len(s1) != len(s2):
        return False

    count = {}

    # Step 2: count characters in first string
    for ch in s1:
        if ch in count:
            count[ch] += 1
        else:
            count[ch] = 1

    # Step 3: subtract using second string
    for ch in s2:
        if ch not in count:
            return False
        count[ch] -= 1

    # Step 4: check all counts are zero
    for value in count.values():
        if value != 0:
            return False

    return True

print(are_anagrams("listen", "silent"))   # True
print(are_anagrams("hello", "world"))     # False
````

````python
def groupAnagrams(strs):
    groups = {}

    for word in strs:
        key = "".join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())

words = ["t", "tea", "tan", "ate", "nat", "bat"]
print(groupAnagrams(words))
````

````python
#Check if one string is a rotation of another.

def is_rotation(s1, s2):
    # Step 1: length must be equal
    if len(s1) != len(s2):
        return False

    # Step 2: check substring
    return s2 in (s1 + s1)
print(is_rotation("abcd", "cdab"))  # True
print(is_rotation("abcd", "acbd"))  # False

````

````python
# Find the longest substring without repeating characters
def longest_unique_substring(s):
    seen = set()
    left = 0
    right = 0
    max_len = 0

    for right in range(len(s)):
        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])
        max_len = max(max_len, right - left + 1)
        right += 1
      

    return max_len
print(longest_unique_substring("abcabcbb"))  # 3
print(longest_unique_substring("bbbb"))      # 1
print(longest_unique_substring("pwwkew"))    # 3

````

````python
def longest_unique_substrings(s):
    seen = set()
    left = 0
    max_len = 0
    result = set()

    for right in range(len(s)):
        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])
        curr_len = right - left + 1
        substr = s[left:right + 1]

        if curr_len > max_len:
            max_len = curr_len
            result = {substr}   # reset
        elif curr_len == max_len:
            result.add(substr)

    return list(result)
print(longest_unique_substrings("abcabcbb"))
````

## Python String Interview – Short Notes (Mid–Senior)

### Time Complexity
- `len(s)`, `s[i]` → **O(1)**
- `s[a:b]` → **O(k)** (new string)
- `s + t` → **O(n + m)**
- `"x" in s` → **O(n)**

### Why concatenation in loops is dangerous
- Strings are immutable → creates new string every time → **O(n²)** overall.

### Naive vs KMP
- Naive: **O(n·m)** worst case
- KMP: **O(n + m)** using prefix table (no backtracking)

### Hashing in strings
- Converts strings to numbers for fast comparison
- Risk: **hash collisions** → must verify actual strings


### Python Traps
- `==` → value equality
It checks whether two variables point to the same object in memory, not whether their values merely look the same.
- `is` → reference equality
- Slicing creates new string (immutability)
- String interning: small/static strings reused
- `''.join(list)` faster than `+=`

---

### Unicode
- `len()` counts code points, not bytes
- UTF-8: variable-length encoding
- One “visible character” ≠ one index

---

### Edge Cases
- Palindrome in stream → two-pointer / rolling hash
- Large strings → chunk processing / streaming
- Same hash ≠ same string
- Regex can be slow, opaque, and overkill

### Why Python string comparison can be faster than expected
- Python first compares **string length**
- Then compares **hash values** (if cached)
- Stops at **first mismatching character**
- Interned strings may compare by **pointer**
→ Average case is very fast; worst case still **O(n)**

---

### Complexity of `split()` and `join()`
- `s.split(delimiter)` → **O(n)**  
  (scans entire string)
- `''.join(list_of_strings)` → **O(total_chars)**  
  (allocates once, copies once)

👉 `join()` is optimal; repeated `+=` is not

### How Python optimizes repeated string concatenation

- CPython applies an optimization for `+=` **only when**:
  - The string has a **single reference**
  - Concatenation happens in a **tight local scope**
- In this case, Python may **resize the string buffer in-place**
  (amortized O(n), similar to list append)

### Where it fails
- When the string has **multiple references**
- When using `s = s + t` instead of `+=`
- When concatenation happens across function boundaries
- When the string escapes scope (e.g., passed to another variable)

### Result
- Best case: near **O(n)**
- Common/worst case: **O(n²)**

Safe rule in interviews: use **`''.join()`**


### `==` vs `is` (for strings)
- `==` → compares **values (content)**
- `is` → compares **object identity (memory address)**
- Two equal strings may be different objects
- `is` may return True due to **string interning** (implementation detail)

---

### Why slicing creates a new string
- Strings are **immutable**
- Slice must return an **independent object**
- Prevents side effects and enables safe sharing
- Cost: **O(k)** time and memory for slice length

---

### Are Python strings mutable? Why this design?
- Strings are **immutable**
- Reasons:
  - Safe sharing & interning
  - Predictable hashing (usable as dict keys)
  - Thread-safety
  - Simpler memory model

Trade-off: immutability costs memory but gains safety & speed


### String Interning
- Interning = storing **one shared copy** of identical strings
- Saves memory and speeds up comparisons (`is`)
- Automatic for:
  - Short strings
  - Identifiers (variable names)
  - Compile-time literals
- Manual interning: `sys.intern()`

---

### Why `"hello" is "hello"` may be True or False
- True when both literals are **interned**
- False when strings are created at **runtime**
- Behavior is **implementation-dependent** (don’t rely on it)

---

### What happens with `s *= 2`
- Creates a **new string** with repeated content
- May reuse buffer **only if** `s` has a single reference
- Time complexity: **O(n)**

---

### `str()` vs `repr()`
- `str()` → user-friendly, readable
- `repr()` → unambiguous, debug-focused
- Goal: `eval(repr(x)) == x` (when possible)


### Why `''.join(list_of_strings)` is faster than `+=`
- `join()`:
  - Computes total length once
  - Allocates memory once
  - Copies characters once → **O(n)**
- `+=` in loops:
  - Reallocates repeatedly
  - Copies old data each time → **O(n²)**

👉 `join()` avoids repeated memory churn

---

### Characters vs Bytes vs Code Points
- **Character**: what a human perceives (e.g., “é”)
- **Code point**: Unicode number (e.g., U+00E9)
- **Byte**: storage unit (8 bits)

---

### Why `len("é")` can differ
- Python string → counts **code points** → `len("é") == 1`
- UTF-8 bytes → `len("é".encode("utf-8")) == 2`
- UTF-16 bytes → often 2 or 4 bytes

---

### UTF-8 vs UTF-16
- UTF-8:
  - Variable-length (1–4 bytes)
  - ASCII-compatible
  - Space-efficient for English
- UTF-16:
  - 2 or 4 bytes per code point
  - Common in some runtimes (e.g., Java)


### Why Unicode indexing ≠ what users see
- Python indexes **code points**, not visual characters
- Many “characters” are:
  - base letter + combining marks
  - emoji sequences (👨‍👩‍👧‍👦)
- One visible symbol may span **multiple code points**

---

### What breaks if you assume 1 char = 1 byte
- String length calculations
- Cursor movement & slicing
- Truncation logic
- File I/O and network protocols
- Text rendering and alignment

---

### How Python stores Unicode strings
- Uses **PEP 393 (Flexible String Representation)**
- Chooses 1, 2, or 4 bytes per code point automatically
- Optimizes memory while preserving fast indexing
- Encoding happens **only** on `.encode()`


### Processing a string too large for memory
- Use **streaming / chunked reading**
- Process line-by-line or fixed-size buffers
- Maintain rolling state (counters, windows, hashes)
- Avoid full materialization

---

### Can two different strings have the same hash?
- Yes → **hash collision**
- Solution:
  - Use hash only as a **filter**
  - Verify with actual string comparison
  - Use stronger / multiple hashes if needed

---

### Comparing strings in constant memory
- Compare **character-by-character**
- Stop at first mismatch
- Uses **O(1)** extra space, **O(n)** time

---

### Pattern matching across chunks
- Keep **overlap buffer** of size `pattern_len - 1`
- Prepend overlap to next chunk
- Use streaming algorithms:
  - KMP
  - Rolling hash (Rabin–Karp)


### KMP, Rabin–Karp, Z-algorithm (what they are)

- **KMP (Knuth–Morris–Pratt)**  
  Uses prefix (LPS) table to avoid rechecking characters  
  Time: **O(n + m)**, Space: **O(m)**

- **Rabin–Karp**  
  Uses rolling hash to compare substrings numerically  
  Average: **O(n + m)**, Worst: **O(n·m)** (collisions)

- **Z-algorithm**  
  Builds Z-array where Z[i] = length of match with prefix  
  Time: **O(n + m)**, good for prefix-based problems

---

### When to choose which?

- **KMP**
  - Single pattern, guaranteed performance
  - Streaming input
  - Collision-free

- **Rabin–Karp**
  - Multiple pattern matching
  - Large text, probabilistic acceptance
  - When hashing is already used

- **Z-algorithm**
  - Prefix analysis
  - Pattern preprocessing
  - Competitive programming style problems

---

### How hashing speeds up string comparison
- Converts strings → fixed-size numbers
- Integer comparison is **O(1)**
- Avoids character-by-character checks
- Used as a **fast pre-check**, not final truth

---

### Risks of hash collisions
- Different strings → same hash
- Causes false matches
- Must:
  - Re-compare actual strings
  - Use double hashing or strong hashes

---

### Why Python hashes change across runs
- Python uses **hash randomization**
- Prevents hash-flooding attacks (security)
- Ensures dict/set safety
- Result: same string → different hash each run


### Designing substring search for very large text

- **Stream the text**
  - Read in fixed-size chunks
  - Never load full text into memory

- **Choose the algorithm**
  - **KMP** → deterministic, no collisions, streaming-safe
  - **Rabin–Karp** → fast average case, good for multiple patterns
  - **Suffix automaton / suffix array** → if text is indexed once, queried many times

- **Handle chunk boundaries**
  - Keep an overlap of `pattern_length - 1`
  - Carry algorithm state across chunks

- **Scale for production**
  - Preprocess patterns once
  - Use parallel chunk processing when possible
  - Cache frequent patterns

- **Real systems**
  - Search engines: inverted indexes + tokenization
  - Log scanning: streaming KMP / rolling hash

👉 Core idea: **stream + linear-time algorithm + boundary awareness**


### Why strings are immutable in Python
- Enables **safe sharing** of string objects
- Allows **string interning**
- Makes strings **hashable** (usable as dict/set keys)
- Improves thread-safety and reasoning about code

---

### Why `join()` is preferred over concatenation
- Allocates memory **once**
- Avoids repeated copying
- Predictable **O(n)** time
- Concatenation in loops leads to **O(n²)**

---

### Trade-offs of immutability
- More memory usage for modifications
- Slicing and updates create new objects
- Requires different patterns (`list + join`)
- Gains: safety, simplicity, and performance predictability


````python
#Longest palindromic substring

def longestPalindrome(s):
    res = ""

    for i in range(len(s)):
        # odd palindrome
        l = r = i
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if r - l + 1 > len(res):
                res = s[l:r+1]
            l -= 1
            r += 1

        # even palindrome
        l, r = i, i + 1
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if r - l + 1 > len(res):
                res = s[l:r+1]
            l -= 1
            r += 1

    return res
print(longestPalindrome("babad"))  # bab / aba
print(longestPalindrome("cbbd"))   # bb
````

````python
def countPalindromicSubstrings(s):
    count = 0

    for i in range(len(s)):

        # odd length palindromes
        l = r = i
        while l >= 0 and r < len(s) and s[l] == s[r]:
            count += 1
            l -= 1
            r += 1

        # even length palindromes
        l, r = i, i + 1
        while l >= 0 and r < len(s) and s[l] == s[r]:
            count += 1
            l -= 1
            r += 1

    return count

print(countPalindromicSubstrings("aaa"))   # 6
print(countPalindromicSubstrings("abc"))   # 3
````

````python
def strStr(haystack, needle):
    if needle == "":
        return 0

    # Step 1: Build LPS array
    lps = [0] * len(needle)
    length = 0
    i = 1

    while i < len(needle):
        if needle[i] == needle[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1

    # Step 2: KMP search
    i = 0  # index for haystack
    j = 0  # index for needle

    while i < len(haystack):
        if haystack[i] == needle[j]:
            i += 1
            j += 1

        if j == len(needle):
            return i - j   # match found

        elif i < len(haystack) and haystack[i] != needle[j]:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

    return -1
print(strStr("abxabcabcaby", "abcaby"))  # 6
print(strStr("hello", "ll"))             # 2
print(strStr("aaaaa", "bba"))            # -1
````

````python
#(aaabbc → a3b2c1)
def compressString(s):
    if not s:
        return ""

    result = ""
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            result += s[i - 1] + str(count)
            count = 1

    # add last character
    result += s[-1] + str(count)

    return result

print(compressString("aaabbc"))   # a3b2c1
print(compressString("a"))        # a1
print(compressString("abc"))      # a1b1c1
````

````python
#Find the longest common prefix among a list of strings.
def longest_common_prefix(words):
    # Step 1: Check empty list
    if not words:
        return ""

    # Step 2: Take first word as base
    prefix = words[0]

    # Step 3: Compare with other words
    for word in words[1:]:
        while not word.startswith(prefix):
            prefix = prefix[:-1]  # remove last character
            if prefix == "":
                return ""

    return prefix

words = ["flower", "flow", "flight"]
print(longest_common_prefix(words))
````

````python
#Minimum Window Substring (pure logic, no modules)
def minWindow(s, t):
    if not s or not t:
        return ""


    t_count = {}
    for ch in t:
        if ch in t_count:
            t_count[ch] += 1
        else:
            t_count[ch] = 1

    window = {}

    have = 0
    need = len(t_count)

    left = 0
    min_len = float("inf")
    result = ""

    for right in range(len(s)):
        ch = s[right]

        # add character to window
        if ch in window:
            window[ch] += 1
        else:
            window[ch] = 1

        # check if requirement met
        if ch in t_count and window[ch] == t_count[ch]:
            have += 1

        # try shrinking window
        while have == need:
            if right - left + 1 < min_len:
                min_len = right - left + 1
                result = s[left:right+1]

            left_char = s[left]
            window[left_char] -= 1

            if left_char in t_count and window[left_char] < t_count[left_char]:
                have -= 1

            left += 1

    return result

s = "ADOBECODEBANC"
t = "ABC"

print(minWindow(s, t))
````

What’s the difference between == and .equals() (or equivalent)?
```
== compares references (same memory object), while .equals() compares the actual character content.
```
How does string interning work?
```
Identical string literals are stored once in a shared pool so multiple references point to the same memory, saving space and speeding comparisons.
```
Strings look simple, but they sit at the intersection of performance, memory, and language design.

````python
from collections import Counter

class Solution:
    def minWindow(self, s: str, t: str) -> str:
        if not s or not t:
            return ""

        need = Counter(t)      # frequency of required characters
        missing = len(t)       # total chars still missing
        left = 0               # left pointer
        start = end = 0        # result window

        for right, ch in enumerate(s):
            if need[ch] > 0:
                missing -= 1
            need[ch] -= 1

            # when all characters are matched
            while missing == 0:
                if end == 0 or right - left < end - start:
                    start, end = left, right + 1

                # shrink window from the left
                need[s[left]] += 1
                if need[s[left]] > 0:
                    missing += 1
                left += 1

        return s[start:end]

s = "ADOBECODEBANC"
t = "ABC"
print(Solution().minWindow(s, t))
````

````python
# Edit Distance (Levenshtein Distance)
# Full runnable program

def edit_distance(a: str, b: str) -> int:
    m, n = len(a), len(b)

    # Create DP table
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    # Base cases
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    # Fill DP table
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(
                    dp[i - 1][j],     # delete
                    dp[i][j - 1],     # insert
                    dp[i - 1][j - 1]  # replace
                )

    return dp[m][n]


def main():
    print("Edit Distance Calculator")
    print("------------------------")

    word1 = input("Enter first string: ")
    word2 = input("Enter second string: ")

    distance = edit_distance(word1, word2)

    print("\nMinimum Edit Distance:", distance)


if __name__ == "__main__":
    main()
````

## How Are Strings Handled in Memory at Runtime?

- Strings are stored as **contiguous memory blocks** (arrays of bytes or characters).
- A string variable usually holds a **reference (pointer)** to that memory, not the actual data itself.

---

## How Does Garbage Collection Interact With Strings?

- When **no references** point to a string, garbage collection frees its memory.
- **Interned strings** may live longer because they are stored in a shared pool.

---

## How Do Immutable Strings Improve Security and Thread Safety?

- Because strings **cannot change**, no thread can modify shared data unexpectedly.
- Sensitive data (paths, class names, keys) **can’t be altered after creation**, reducing security risks.

---

## How Would You Design a Text Search Engine or Autocomplete System?

Use **indexes**, not raw string scanning:

- **Inverted Index** → Used in search engines
- **Trie / Prefix Tree** → Used for autocomplete

This avoids scanning every string on every query.

## Classic Curveball Questions 

### Why Are Strings Immutable in Java/Python?

- Enables **string interning**
- Makes strings **hash-safe** (usable as dictionary/map keys)
- Prevents **accidental or malicious modification**
