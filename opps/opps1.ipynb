| Method Type         | Definition                                                 | Example                             |
| ------------------- | ---------------------------------------------------------- | ----------------------------------- |
| **Instance Method** | Works with object data (instance variables). Uses `self`.  | `def display(self):`                |
| **Class Method**    | Works with class variables. Uses `@classmethod` and `cls`. | `@classmethod`<br>`def show(cls):`  |

````python
#1. Instance Method
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):   # Instance Method
        print(self.name)

s1 = Student("Alice")
s1.display()
````

````python
#Class Method


class Student:
    school = "ABC School"

    @classmethod
    def show(cls):      # Class Method
        print(cls.school)

Student.show()
````

````python
#Static Method

#Definition: Does not use object or class data. Uses @staticmethod.

class Student:
    @staticmethod
    def add(a, b):      # Static Method
        print(a + b)

Student.add(10, 20)
````

## Encapsulation

Encapsulation is the process of wrapping data (attributes) and methods (functions) into a single class. It protects the object's data from direct access by restricting unauthorized modification. Data is usually accessed or modified through public methods such as getters and setters.

### Public Members

Definition: Public members are attributes or methods that can be accessed directly from anywhere inside or outside the class. They have no underscore (_) before their name.

### Protected Members (_variable)

Definition: Protected members are intended to be accessed only within the class and its subclasses. They are prefixed with a single underscore (_), but they can still be accessed from outside by convention.

### Private Members (__variable)

Definition: Private members are intended to be accessed only within the same class. They are prefixed with double underscores (__), making direct access from outside the class difficult through name mangling.

### Getter Methods

Definition: Getter methods are used to safely read or retrieve the value of a private or protected attribute without accessing it directly.

### Setter Methods

Definition: Setter methods are used to safely update or modify the value of a private or protected attribute, often with validation before assigning the new value.

````python
class Student:
    def __init__(self, name, age, marks):
        self.name = name        # Public
        self._age = age         # Protected
        self.__marks = marks    # Private

    # Getter Method
    def get_marks(self):
        return self.__marks

    # Setter Method
    def set_marks(self, marks):
        self.__marks = marks


s1 = Student("Alice", 20, 85)

# Public Member
print(s1.name)

# Protected Member
print(s1._age)

# Getter Method
print(s1.get_marks())

# Setter Method
s1.set_marks(95)
print(s1.get_marks())
````

## Abstraction

Abstraction is an Object-Oriented Programming (OOP) concept that focuses on hiding unnecessary implementation details and showing only the essential features of an object. It allows programmers to define what an object does without exposing how it does it. This simplifies complex systems, improves code readability, and makes programs easier to maintain and extend. In Python, abstraction is mainly achieved using abstract classes and abstract methods.

Abstract Classes

An abstract class is a special type of class that cannot be instantiated directly. It acts as a blueprint for other classes by defining common methods and attributes that derived classes must implement. An abstract class may contain both abstract methods (methods without implementation) and concrete methods (methods with implementation). In Python, abstract classes are created using the ABC class and the @abstractmethod decorator from the abc module. Any class inheriting from an abstract class must provide implementations for all abstract methods before objects of that class can be created.

Abstract Method

An abstract method is a method declared inside an abstract class without an implementation. Every child class must implement the abstract method; otherwise, its object cannot be created.

````python
from abc import ABC, abstractmethod

# Abstract Class
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


# Child Class
class Dog(Animal):

    def sound(self):
        print("Dog barks")


# Creating object
dog = Dog()
dog.sound()
````

```
ABC stands for Abstract Base Class.
Animal is an abstract class, so you cannot create its object.
@abstractmethod makes the sound() method compulsory for all child classes.
Dog and Cat inherit from Animal and provide their own implementation of sound().
If a child class does not implement sound(), Python raises an error when you try to create its object.


| Term              | Type                | Purpose                                                                                                                              | Example                                 |
| ----------------- | ------------------- | ------------------------------------------------------------------------------------------------------------------------------------ | --------------------------------------- |
| `ABC`             | Abstract Base Class | Used to create an abstract class. A class that inherits from `ABC` can contain abstract methods and cannot be instantiated directly. | `class Animal(ABC):`                    |
| `@abstractmethod` | Decorator           | Marks a method as abstract. Every child class must implement this method before an object can be created.                            | `@abstractmethod`<br>`def sound(self):` |


````python
from abc import ABC, abstractmethod

# Abstract Class
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


# Child Class
class Dog(Animal):

    def sound(self):
        print("Dog barks")


# Child Class
class Cat(Animal):

    def sound(self):
        print("Cat meows")


# Creating objects
dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

## Inheritance

Inheritance is an Object-Oriented Programming (OOP) concept that allows one class (called the child or derived class) to inherit the properties and methods of another class (called the parent or base class). It promotes code reusability, reduces duplication, and makes programs easier to maintain by allowing child classes to extend or modify the functionality of the parent class.

## Single Inheritance

Single inheritance is a type of inheritance in which one child class inherits from one parent class. The child class gains access to all accessible attributes and methods of the parent class and can also define its own additional features. This is the simplest and most commonly used form of inheritance.

## Multiple Inheritance

Multiple inheritance is a type of inheritance in which one child class inherits from two or more parent classes. The child class can access the attributes and methods of all parent classes, allowing it to combine functionalities from multiple sources into a single class.

## Multilevel Inheritance

Multilevel inheritance is a type of inheritance where a class inherits from another derived class. In this structure, inheritance forms a chain, such as Grandparent → Parent → Child, allowing the child class to inherit features from both the parent and grandparent classes.

## Hierarchical Inheritance

Hierarchical inheritance is a type of inheritance in which multiple child classes inherit from the same parent class. Each child class shares the common features of the parent class while also having its own unique properties and methods.

## Hybrid Inheritance

Hybrid inheritance is a combination of two or more types of inheritance, such as single, multiple, and multilevel inheritance, within the same program. It is used to model complex relationships between classes. In Python, hybrid inheritance is supported because the language allows multiple inheritance.

````python
# Single Inheritance
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

dog = Dog()
dog.sound()
dog.bark()
````

````python
# Parent class
class Grandparent:
    def grandparent_skill(self):
        print("Grandparent knows farming.")


# Child of Grandparent (Multilevel)
class Parent(Grandparent):
    def parent_skill(self):
        print("Parent knows teaching.")


# Another Parent class
class Sports:
    def sports_skill(self):
        print("Sports class teaches football.")


# Child inherits from Parent and Sports (Multiple)
class Child(Parent, Sports):
    def child_skill(self):
        print("Child knows programming.")


# Create object
c = Child()

# Call methods
c.grandparent_skill()
c.parent_skill()
c.sports_skill()
c.child_skill()
````
