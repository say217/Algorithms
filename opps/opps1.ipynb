It defines the attributes (data) and methods (functions) that objects will have.

````python
class Student:
    def __init__(self, name , age):
        self.name = name
        self.age = age
        
    def display(self):
            print(self.name, self.age)
            
````

It is created from a class and can access its attributes and methods.

````python
s1 = Student("Alice", 20)
s2 = Student("Bob", 22)

s1.display()
s2.display()
````

Attributes (Variables)

Example:
```python

```
Methods (Functions)

Definition: Methods are functions inside a class that define the object's behavior.

Example:
```python
class Student:
    def display(self):        # Method
        print("Hello, Student!")

| Method Type         | Definition                                                 | Example                             |
| ------------------- | ---------------------------------------------------------- | ----------------------------------- |
| **Instance Method** | Works with object data (instance variables). Uses `self`.  | `def display(self):`                |
| **Class Method**    | Works with class variables. Uses `@classmethod` and `cls`. | `@classmethod`<br>`def show(cls):`  |
| **Static Method**   | Does not use object or class data. Uses `@staticmethod`.   | `@staticmethod`<br>`def add(a, b):` |


````python
#1. Instance Method
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):   # Instance Method
        print(self.name)

s1 = Student("Alice")
s1.display()
````

````python
#Class Method


class Student:
    school = "ABC School"

    @classmethod
    def show(cls):      # Class Method
        print(cls.school)

Student.show()
````

````python
#Static Method

#Definition: Does not use object or class data. Uses @staticmethod.

class Student:
    @staticmethod
    def add(a, b):      # Static Method
        print(a + b)

Student.add(10, 20)
````

# Types of Attributes in Python OOP

| Attribute Type         | Definition                                   | Example                 |
| ---------------------- | -------------------------------------------- | ----------------------- |
| **Instance Attribute** | Belongs to an object. Created using `self`.  | `self.name = "Alice"`   |
| **Class Attribute**    | Belongs to the class. Shared by all objects. | `school = "ABC School"` |


````python
#. Instance Attribute


class Student:
    def __init__(self, name):
        self.name = name    # Instance Attribute

s1 = Student("Alice")
print(s1.name)
````

````python
#Class Attribute

class Student:
    school = "ABC School"   # Class Attribute

s1 = Student()
s2 = Student()

print(s1.school)
print(s2.school)
````

## Encapsulation

Encapsulation is the process of wrapping data (attributes) and methods (functions) into a single class. It protects the object's data from direct access by restricting unauthorized modification. Data is usually accessed or modified through public methods such as getters and setters.

### Public Members

Definition: Public members are attributes or methods that can be accessed directly from anywhere inside or outside the class. They have no underscore (_) before their name.

### Protected Members (_variable)

Definition: Protected members are intended to be accessed only within the class and its subclasses. They are prefixed with a single underscore (_), but they can still be accessed from outside by convention.

### Private Members (__variable)

Definition: Private members are intended to be accessed only within the same class. They are prefixed with double underscores (__), making direct access from outside the class difficult through name mangling.

### Getter Methods

Definition: Getter methods are used to safely read or retrieve the value of a private or protected attribute without accessing it directly.

### Setter Methods

Definition: Setter methods are used to safely update or modify the value of a private or protected attribute, often with validation before assigning the new value.

````python
class Student:
    def __init__(self, name, age, marks):
        self.name = name        # Public
        self._age = age         # Protected
        self.__marks = marks    # Private

    # Getter Method
    def get_marks(self):
        return self.__marks

    # Setter Method
    def set_marks(self, marks):
        self.__marks = marks


s1 = Student("Alice", 20, 85)

# Public Member
print(s1.name)

# Protected Member
print(s1._age)

# Getter Method
print(s1.get_marks())

# Setter Method
s1.set_marks(95)
print(s1.get_marks())
````

### 1. Public Member (`self.name`)

```python
self.name = name
print(s1.name)
```

- `name` is a **public attribute**.
- It can be accessed directly from outside the class.

---

### 2. Protected Member (`self._age`)

```python
self._age = age
print(s1._age)
```

- `_age` is a **protected attribute**.
- It should only be used inside the class or subclasses, but Python still allows direct access.

---

### 3. Private Member (`self.__marks`)

```python
self.__marks = marks
```

- `__marks` is a **private attribute**.
- It cannot be accessed directly using `s1.__marks`.

---

### 4. Getter Method

```python
def get_marks(self):
    return self.__marks

print(s1.get_marks())
```

- Returns the value of the private attribute `__marks`.
- Used to read private data safely.

---

### 5. Setter Method

```python
def set_marks(self, marks):
    self.__marks = marks

s1.set_marks(95)
```

- Updates the value of the private attribute `__marks`.
- Used to modify private data safely.
