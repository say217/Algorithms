## Aggregation

Aggregation is an object-oriented programming (OOP) concept in which one class uses an object of another class, but the contained object can exist independently of the main class. It represents a weak Has-A relationship, meaning the lifetime of the contained object does not depend on the lifetime of the container object. Aggregation promotes code reusability and loose coupling because the same object can be shared by multiple classes.

## Weak Has-A Relationship

A Weak Has-A relationship means one object uses another object, but it does not own it. Even if the main object is destroyed, the other object continues to exist independently. For example, a Student has a Teacher, but deleting the student does not delete the teacher. Similarly, a Department has Professors, but professors can exist even if the department object is removed.

````python
class Teacher:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Teacher:", self.name)


class Student:
    def __init__(self, name, teacher):
        self.name = name
        self.teacher = teacher   # Aggregation

    def display(self):
        print("Student:", self.name)
        self.teacher.display()


# Teacher object created independently
t1 = Teacher("Mr. Sharma")

# Student uses the existing Teacher object
s1 = Student("Rahul", t1)

s1.display()
````

## Association

Association is an object-oriented programming (OOP) relationship where two or more independent objects communicate or interact with each other to perform a task. Unlike composition and aggregation, neither object owns the other. Both objects can exist independently and simply know about or use each other. Association represents a general "uses-a" relationship and can be One-to-One, One-to-Many, or Many-to-Many.

## One-to-One Association

A One-to-One Association means one object is associated with exactly one other object, and vice versa. Both objects are independent and can exist without each other.

````python
class Passport:
    def __init__(self, number):
        self.number = number


class Person:
    def __init__(self, name, passport):
        self.name = name
        self.passport = passport


p = Passport("P123456")
person = Person("Rahul", p)

print("Person :", person.name)
print("Passport :", person.passport.number)
````

## One-to-Many Association

A One-to-Many Association means one object is associated with multiple objects, while each of those objects is associated with only one main object.

````python
class Student:
    def __init__(self, name):
        self.name = name


class Teacher:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def display(self):
        print("Teacher:", self.name)
        print("Students:")
        for student in self.students:
            print(student.name)


s1 = Student("Rahul")
s2 = Student("Anita")
s3 = Student("Riya")

t = Teacher("Mr. Sharma")

t.add_student(s1)
t.add_student(s2)
t.add_student(s3)

t.display()
````

## Many-to-Many Association

A Many-to-Many Association means multiple objects of one class are associated with multiple objects of another class. Both classes maintain references to each other.

````python
class Student:
    def __init__(self, name):
        self.name = name
        self.courses = []

    def enroll(self, course):
        self.courses.append(course)


class Course:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)


s1 = Student("Rahul")
s2 = Student("Anita")

c1 = Course("Python")
c2 = Course("AI")

# Associations
s1.enroll(c1)
s1.enroll(c2)
s2.enroll(c1)

c1.add_student(s1)
c1.add_student(s2)
c2.add_student(s1)

print("Rahul Courses:")
for course in s1.courses:
    print(course.name)

print("\nPython Students:")
for student in c1.students:
    print(student.name)
````

| Association Type | Description                                           | Example            |
| ---------------- | ----------------------------------------------------- | ------------------ |
| **One-to-One**   | One object is associated with one other object        | Person ↔ Passport  |
| **One-to-Many**  | One object is associated with multiple objects        | Teacher → Students |
| **Many-to-Many** | Multiple objects are associated with multiple objects | Students ↔ Courses |


## Object Lifecycle

The object lifecycle in Python describes the complete journey of an object from the moment it is created until it is removed from memory. The lifecycle consists of four main stages: Creation, Reference Counting, Garbage Collection, and Deletion. Python automatically manages memory during these stages, allowing programmers to focus on writing code rather than manually allocating and freeing memory.

## Creation

Creation is the first stage of an object's lifecycle. An object is created when a class is instantiated using its constructor. At the time of creation, Python allocates memory for the object and automatically calls the __init__() constructor to initialize its instance variables. After initialization, the object becomes ready for use.

````python
class Student:
    def __init__(self, name):
        self.name = name
        print("Object Created")

s1 = Student("Rahul")
print(s1.name)
````

## Reference Counting

Reference counting is Python's primary memory management technique. Every object maintains a count of how many references (variables) point to it. Whenever a new variable refers to the object, the reference count increases. When a reference is removed or reassigned, the reference count decreases. If the reference count becomes zero, the object becomes eligible for deletion because no part of the program can access it anymore.

sys.getrefcount() usually shows one extra reference because the function itself temporarily references the object while counting.

````python
import sys

class Student:
    pass

s1 = Student()

print("Reference Count:", sys.getrefcount(s1))

s2 = s1      # Another reference

print("Reference Count:", sys.getrefcount(s1))
````

## Garbage Collection

Garbage Collection is the automatic process used by Python to free memory occupied by objects that are no longer needed. Although reference counting handles most objects, it cannot clean up objects involved in circular references (objects referring to each other). Python's Garbage Collector (GC) detects these unreachable objects and removes them automatically, preventing memory leaks.

````python
import gc

class Node:
    pass

a = Node()
b = Node()

# Circular reference
a.next = b
b.next = a

# Remove references
del a
del b

# Run Garbage Collector
gc.collect()

print("Garbage Collection Completed")
````

## Deletion

Deletion is the final stage of an object's lifecycle. When an object has no remaining references and is no longer needed, Python removes it from memory. Before the object is destroyed, Python calls the __del__() destructor (if it is defined) to perform cleanup operations such as closing files or releasing resourc

````python
class Student:
    def __init__(self, name):
        self.name = name
        print("Object Created")

    def __del__(self):
        print("Object Deleted")

s1 = Student("Rahul")

del s1
````

| Stage                  | Description                                                                     |
| ---------------------- | ------------------------------------------------------------------------------- |
| **Creation**           | Python allocates memory and calls `__init__()` when an object is created.       |
| **Reference Counting** | Python tracks how many variables refer to an object.                            |
| **Garbage Collection** | Removes unreachable objects, including those with circular references.          |
| **Deletion**           | When no references remain, Python destroys the object and may call `__del__()`. |


````python
import gc

class Student:

    def __init__(self, name):
        self.name = name
        print("Object Created")

    def __del__(self):
        print("Object Deleted")


# Creation
s1 = Student("Rahul")

# Reference Counting
s2 = s1

print("Two references exist")

# Remove one reference
del s2

print("One reference remains")

# Remove last reference
del s1

# Force Garbage Collection
gc.collect()
````

## Properties

Properties in Python provide a controlled way to access, modify, and delete an object's attributes while keeping the syntax simple. Instead of directly accessing instance variables, properties allow methods to behave like attributes. The @property decorator is used to create a getter, which returns the value of an attribute. The setter (@property_name.setter) is used to update the value and can include validation before assigning it. The deleter (@property_name.deleter) is used to define what happens when an attribute is deleted using the del keyword. Properties support the principle of encapsulation, allowing programmers to protect object data, perform validation, and control how attributes are accessed without changing the way users interact with the object.

| Decorator                | Purpose | Called When                            |
| ------------------------ | ------- | -------------------------------------- |
| `@property`              | Getter  | Reading an attribute (`obj.name`)      |
| `@property_name.setter`  | Setter  | Assigning a value (`obj.name = value`) |
| `@property_name.deleter` | Deleter | Deleting an attribute (`del obj.name`) |


````python
class Student:

    def __init__(self, name):
        self._name = name

    # Getter
    @property
    def name(self):
        print("Getter called")
        return self._name

    # Setter
    @name.setter
    def name(self, value):
        print("Setter called")
        self._name = value

    # Deleter
    @name.deleter
    def name(self):
        print("Deleter called")
        del self._name


# Create object
s = Student("Rahul")

# Getter
print(s.name)

# Setter
s.name = "Anita"

# Getter
print(s.name)

# Deleter
del s.name
````

## Decorators in OOP

Decorators in Python are special functions that modify or extend the behavior of methods without changing their actual code. In object-oriented programming (OOP), decorators are commonly used to define special types of methods such as properties, class methods, and static methods. A decorator is written using the @ symbol before a method definition. Two of the most commonly used decorators are @property, which provides controlled access to object attributes, and @classmethod, which defines methods that operate on the class rather than individual objects.

## @property

The @property decorator is used to create a getter method that can be accessed like a normal attribute. It provides a controlled way to read an object's data without directly exposing the internal variables. When an attribute is accessed, Python automatically calls the method decorated with @property. It is commonly used to implement encapsulation, perform validation, or compute values dynamically while keeping the syntax simple.

````python
class Student:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name


# Create object
s = Student("Rahul")

# Access property
print(s.name)
````

## @classmethod

The @classmethod decorator is used to define a method that belongs to the class rather than an individual object. A class method receives cls as its first parameter, where cls refers to the class itself. Class methods can access and modify class variables but cannot directly access instance variables unless an object is provided. They are commonly used to create alternative constructors, modify shared class data, or perform operations related to the class as a whole.

````python
class Student:

    school = "ABC School"

    @classmethod
    def display_school(cls):
        print("School Name:", cls.school)


# Call class method
Student.display_school()
````

## @staticmethod

The @staticmethod decorator is used to define a method that belongs to a class but does not depend on either the object (self) or the class (cls). A static method behaves like a normal function placed inside a class for better organization. Since it does not access instance variables or class variables directly, it does not require self or cls as its first parameter. Static methods are commonly used for utility functions, helper methods, mathematical calculations, data validation, or any operation that is logically related to the class but does not need access to the object's or class's data.

````python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b


# Call using class
print(Calculator.add(10, 20))

# Call using object
obj = Calculator()
print(obj.add(30, 40))
````

## Custom Decorators

A custom decorator is a user-defined function that modifies or extends the behavior of another function or method without changing its original code. Custom decorators are useful for adding common functionality such as logging, authentication, timing, input validation, or access control. A decorator takes a function as an argument, defines a wrapper function that adds extra behavior, and returns the wrapper. The decorated function is automatically passed to the decorator when the @decorator_name syntax is used.

````python
# Custom Decorator
def message(func):
    def wrapper(self):
        print("Before calling function")
        func(self)
        print("After calling function")
    return wrapper


class Student:

    @message
    def display(self):
        print("Welcome to Python OOP")


# Create object
s = Student()

# Call method
s.display()
````

- *args collects any number of positional arguments into a tuple, allowing a function to accept a variable number of arguments without knowing them in advance.

- **kwargs collects any number of keyword arguments into a dictionary, making functions and decorators flexible enough to accept and forward any named parameters.

````python
def message(func):
    def wrapper(*args, **kwargs):
        print("Before calling function")
        result = func(*args, **kwargs)
        print("After calling function")
        return result
    return wrapper


class Student:

    @message
    def display(self):
        print("Welcome to Python OOP")


s = Student()
s.display()
````

## Abstract Base Classes (ABC)

An Abstract Base Class (ABC) is a special type of class that acts as a blueprint for other classes. It cannot be instantiated (you cannot create an object from it directly). An abstract base class defines common methods that every child class must implement. In Python, an abstract base class is created by inheriting from ABC, which is imported from the abc module. ABCs enforce a common structure among related classes and support abstraction in object-oriented programming.

## Abstract Class

An abstract class is a class that contains one or more abstract methods. It is designed to be inherited by other classes rather than used to create objects directly. An abstract class may contain both abstract methods (without implementation) and normal methods (with implementation). Its main purpose is to provide a common interface that all subclasses must follow.

````python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
````

## Abstract Method

An abstract method is a method declared using the @abstractmethod decorator. It contains no implementation (or only pass) in the abstract class. Every child class must override and implement all abstract methods; otherwise, the child class also becomes abstract and cannot be instantiated.

| Term                            | Description                                                                             |
| ------------------------------- | --------------------------------------------------------------------------------------- |
| **ABC (`from abc import ABC`)** | Base class used to create abstract classes.                                             |
| **Abstract Class**              | A class that cannot be instantiated and serves as a blueprint for subclasses.           |
| **Abstract Method**             | A method declared with `@abstractmethod` that must be implemented by every child class. |


````python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


r = Rectangle(10, 5)
print("Area:", r.area())
````

## Interfaces (Python Style)

In Python, there is no separate interface keyword like in Java or C#. Instead, abstract classes from the abc module are used to define interfaces. An interface specifies a set of methods that a class must implement, but it does not define how those methods should work. Any class that inherits from the interface must provide implementations for all abstract methods; otherwise, it cannot be instantiated. This ensures that different classes follow the same structure while allowing each class to have its own implementation.

Using Abstract Classes to Define Interfaces

Python uses the ABC class and the @abstractmethod decorator to create an interface. The interface declares the required methods, and every child class must override those methods with its own implementation.

```bash
Advantages of Interfaces
Defines a common structure for related classes.
Enforces implementation of required methods.
Supports abstraction and polymorphism.
Improves code consistency and maintainability.
Makes programs easier to extend and reuse.

````python
from abc import ABC, abstractmethod

# Interface
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


# Child Class 1
class Dog(Animal):

    def sound(self):
        print("Dog barks")


# Child Class 2
class Cat(Animal):

    def sound(self):
        print("Cat meows")


# Create objects
d = Dog()
c = Cat()

d.sound()
c.sound()
````

## Operator Overloading

Operator overloading is an object-oriented programming (OOP) feature that allows programmers to define how Python operators behave when used with objects of a user-defined class. Python provides special (magic/dunder) methods such as __add__(), __sub__(), and __eq__() to overload operators. When an operator is used between two objects, Python automatically calls the corresponding special method. This makes user-defined objects behave like built-in data types and improves code readability and flexibility.

````python
class Number:

    def __init__(self, value):
        self.value = value

    # + Operator
    def __add__(self, other):
        return self.value + other.value

    # - Operator
    def __sub__(self, other):
        return self.value - other.value

    # * Operator
    def __mul__(self, other):
        return self.value * other.value

    # == Operator
    def __eq__(self, other):
        return self.value == other.value

    # < Operator
    def __lt__(self, other):
        return self.value < other.value

    # > Operator
    def __gt__(self, other):
        return self.value > other.value


n1 = Number(20)
n2 = Number(10)
n3 = Number(20)

print("Addition       :", n1 + n2)
print("Subtraction    :", n1 - n2)
print("Multiplication :", n1 * n2)
print("Equal          :", n1 == n3)
print("Less Than      :", n1 < n2)
print("Greater Than   :", n1 > n2)
````

## Iterators and Iterable Objects

An iterable object is any object that can be traversed one element at a time, such as a list, tuple, string, set, or dictionary. An iterable does not produce elements directly; instead, it returns an iterator. An iterator is an object that keeps track of the current position during iteration and returns one element at a time until no elements remain. In Python, iteration is performed using the special methods __iter__() and __next__(). These methods make objects compatible with for loops and the next() function.

## __iter__()

The __iter__() method is automatically called when iteration begins, such as in a for loop or by using the iter() function. It returns an iterator object, which is responsible for producing the elements one by one. In many custom iterators, __iter__() simply returns self.

## __next__()

The __next__() method returns the next element in the sequence each time it is called. When there are no more elements to return, it raises the StopIteration exception, which tells Python that the iteration has finished.

## Custom Iterator

A custom iterator is a user-defined class that implements both the __iter__() and __next__() methods. This allows programmers to define their own iteration logic instead of relying on Python's built-in iterables. Custom iterators are useful when generating sequences, processing data step by step, or controlling how iteration should occur.

````python
class Numbers:

    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    # Return iterator object
    def __iter__(self):
        return self

    # Return next value
    def __next__(self):
        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value
        raise StopIteration


# Create iterator
num = Numbers(5)

# Iterate using for loop
for i in num:
    print(i)
````

## Callable Objects

A callable object is an object that can be called like a function using parentheses (). In Python, an object becomes callable by implementing the special __call__() method. When parentheses are placed after an object (for example, obj()), Python automatically invokes the object's __call__() method. Callable objects are useful when an object needs to behave like a function while still maintaining its own data (state). They are commonly used in decorators, machine learning models, and classes that perform repeated operations.

__call__()

The __call__() method is a special (magic/dunder) method that makes an object callable. It is automatically executed whenever the object is called using parentheses (). The __call__() method can accept arguments, perform computations, and return values just like a normal function.

````python
class Counter:

    def __init__(self):
        self.count = 0

    def __call__(self):
        self.count += 1
        print("Called", self.count, "times")


counter = Counter()

counter()
counter()
counter()
````

````python
#: Callable Object with Arguments
class Calculator:

    def __call__(self, a, b):
        return a + b


calc = Calculator()

print(calc(10, 20))
````
