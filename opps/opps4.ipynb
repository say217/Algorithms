## Aggregation

Aggregation is an object-oriented programming (OOP) concept in which one class uses an object of another class, but the contained object can exist independently of the main class. It represents a weak Has-A relationship, meaning the lifetime of the contained object does not depend on the lifetime of the container object. Aggregation promotes code reusability and loose coupling because the same object can be shared by multiple classes.

## Weak Has-A Relationship

A Weak Has-A relationship means one object uses another object, but it does not own it. Even if the main object is destroyed, the other object continues to exist independently. For example, a Student has a Teacher, but deleting the student does not delete the teacher. Similarly, a Department has Professors, but professors can exist even if the department object is removed.

```python
class Teacher:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Teacher:", self.name)


class Student:
    def __init__(self, name, teacher):
        self.name = name
        self.teacher = teacher   # Aggregation

    def display(self):
        print("Student:", self.name)
        self.teacher.display()


# Teacher object created independently
t1 = Teacher("Mr. Sharma")

# Student uses the existing Teacher object
s1 = Student("Rahul", t1)

s1.display()
```

## Association

Association is an object-oriented programming (OOP) relationship where two or more independent objects communicate or interact with each other to perform a task. Unlike composition and aggregation, neither object owns the other. Both objects can exist independently and simply know about or use each other. Association represents a general "uses-a" relationship and can be One-to-One, One-to-Many, or Many-to-Many.

## One-to-One Association

A One-to-One Association means one object is associated with exactly one other object, and vice versa. Both objects are independent and can exist without each other.

```python
class Passport:
    def __init__(self, number):
        self.number = number


class Person:
    def __init__(self, name, passport):
        self.name = name
        self.passport = passport


p = Passport("P123456")
person = Person("Rahul", p)

print("Person :", person.name)
print("Passport :", person.passport.number)
```

## One-to-Many Association

A One-to-Many Association means one object is associated with multiple objects, while each of those objects is associated with only one main object.

```python
class Student:
    def __init__(self, name):
        self.name = name


class Teacher:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def display(self):
        print("Teacher:", self.name)
        print("Students:")
        for student in self.students:
            print(student.name)


s1 = Student("Rahul")
s2 = Student("Anita")
s3 = Student("Riya")

t = Teacher("Mr. Sharma")

t.add_student(s1)
t.add_student(s2)
t.add_student(s3)

t.display()
```

## Many-to-Many Association

A Many-to-Many Association means multiple objects of one class are associated with multiple objects of another class. Both classes maintain references to each other.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.courses = []

    def enroll(self, course):
        self.courses.append(course)


class Course:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)


s1 = Student("Rahul")
s2 = Student("Anita")

c1 = Course("Python")
c2 = Course("AI")

# Associations
s1.enroll(c1)
s1.enroll(c2)
s2.enroll(c1)

c1.add_student(s1)
c1.add_student(s2)
c2.add_student(s1)

print("Rahul Courses:")
for course in s1.courses:
    print(course.name)

print("\nPython Students:")
for student in c1.students:
    print(student.name)
```

| Association Type | Description                                           | Example            |
| ---------------- | ----------------------------------------------------- | ------------------ |
| **One-to-One**   | One object is associated with one other object        | Person ↔ Passport  |
| **One-to-Many**  | One object is associated with multiple objects        | Teacher → Students |
| **Many-to-Many** | Multiple objects are associated with multiple objects | Students ↔ Courses |


## Object Lifecycle

The object lifecycle in Python describes the complete journey of an object from the moment it is created until it is removed from memory. The lifecycle consists of four main stages: Creation, Reference Counting, Garbage Collection, and Deletion. Python automatically manages memory during these stages, allowing programmers to focus on writing code rather than manually allocating and freeing memory.

## Creation

Creation is the first stage of an object's lifecycle. An object is created when a class is instantiated using its constructor. At the time of creation, Python allocates memory for the object and automatically calls the __init__() constructor to initialize its instance variables. After initialization, the object becomes ready for use.

```python
class Student:
    def __init__(self, name):
        self.name = name
        print("Object Created")

s1 = Student("Rahul")
print(s1.name)
```

## Reference Counting

Reference counting is Python's primary memory management technique. Every object maintains a count of how many references (variables) point to it. Whenever a new variable refers to the object, the reference count increases. When a reference is removed or reassigned, the reference count decreases. If the reference count becomes zero, the object becomes eligible for deletion because no part of the program can access it anymore.

sys.getrefcount() usually shows one extra reference because the function itself temporarily references the object while counting.

```python
import sys

class Student:
    pass

s1 = Student()

print("Reference Count:", sys.getrefcount(s1))

s2 = s1      # Another reference

print("Reference Count:", sys.getrefcount(s1))
```

## Garbage Collection

Garbage Collection is the automatic process used by Python to free memory occupied by objects that are no longer needed. Although reference counting handles most objects, it cannot clean up objects involved in circular references (objects referring to each other). Python's Garbage Collector (GC) detects these unreachable objects and removes them automatically, preventing memory leaks.

```python
import gc

class Node:
    pass

a = Node()
b = Node()

# Circular reference
a.next = b
b.next = a

# Remove references
del a
del b

# Run Garbage Collector
gc.collect()

print("Garbage Collection Completed")
```

## Deletion

Deletion is the final stage of an object's lifecycle. When an object has no remaining references and is no longer needed, Python removes it from memory. Before the object is destroyed, Python calls the __del__() destructor (if it is defined) to perform cleanup operations such as closing files or releasing resourc

```python
class Student:
    def __init__(self, name):
        self.name = name
        print("Object Created")

    def __del__(self):
        print("Object Deleted")

s1 = Student("Rahul")

del s1
```

| Stage                  | Description                                                                     |
| ---------------------- | ------------------------------------------------------------------------------- |
| **Creation**           | Python allocates memory and calls `__init__()` when an object is created.       |
| **Reference Counting** | Python tracks how many variables refer to an object.                            |
| **Garbage Collection** | Removes unreachable objects, including those with circular references.          |
| **Deletion**           | When no references remain, Python destroys the object and may call `__del__()`. |


```python
import gc

class Student:

    def __init__(self, name):
        self.name = name
        print("Object Created")

    def __del__(self):
        print("Object Deleted")


# Creation
s1 = Student("Rahul")

# Reference Counting
s2 = s1

print("Two references exist")

# Remove one reference
del s2

print("One reference remains")

# Remove last reference
del s1

# Force Garbage Collection
gc.collect()
```

## Properties

Properties in Python provide a controlled way to access, modify, and delete an object's attributes while keeping the syntax simple. Instead of directly accessing instance variables, properties allow methods to behave like attributes. The @property decorator is used to create a getter, which returns the value of an attribute. The setter (@property_name.setter) is used to update the value and can include validation before assigning it. The deleter (@property_name.deleter) is used to define what happens when an attribute is deleted using the del keyword. Properties support the principle of encapsulation, allowing programmers to protect object data, perform validation, and control how attributes are accessed without changing the way users interact with the object.

| Decorator                | Purpose | Called When                            |
| ------------------------ | ------- | -------------------------------------- |
| `@property`              | Getter  | Reading an attribute (`obj.name`)      |
| `@property_name.setter`  | Setter  | Assigning a value (`obj.name = value`) |
| `@property_name.deleter` | Deleter | Deleting an attribute (`del obj.name`) |


```python
class Student:

    def __init__(self, name):
        self._name = name

    # Getter
    @property
    def name(self):
        print("Getter called")
        return self._name

    # Setter
    @name.setter
    def name(self, value):
        print("Setter called")
        self._name = value

    # Deleter
    @name.deleter
    def name(self):
        print("Deleter called")
        del self._name


# Create object
s = Student("Rahul")

# Getter
print(s.name)

# Setter
s.name = "Anita"

# Getter
print(s.name)

# Deleter
del s.name
```
