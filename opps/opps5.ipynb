## Threads in Python

A thread is the smallest unit of execution inside a process. Multiple threads can run within the same program and share the same memory. Threads are useful when a program needs to perform multiple tasks at the same time, such as downloading files while updating the user interface or handling multiple client requests. Python provides thread support through the built-in threading module.

| Thread Type                  | Description                                                                |
| ---------------------------- | -------------------------------------------------------------------------- |
| **Main Thread**              | The first thread that starts when a Python program begins execution.       |
| **User (Non-Daemon) Thread** | A normal thread that completes its task before the program exits.          |
| **Daemon Thread**            | A background thread that automatically stops when all user threads finish. |


## Main Thread

The main thread is automatically created when a Python program starts. It is responsible for executing the main code of the program. Any additional threads are usually created by the main thread.

```python
import threading

def show():
    print("Current Thread:", threading.current_thread().name)

print("Main program started")
show()
print("Main program ended")
```

## User (Non-Daemon) Thread

A user thread (also called a non-daemon thread) performs normal tasks. Python waits for all user threads to finish before terminating the program.

```python
import threading
import time

def task():
    for i in range(3):
        print("User Thread Running")
        time.sleep(1)

t = threading.Thread(target=task)
t.start()

print("Main Thread Finished")
```

## Daemon Thread

A daemon thread runs in the background to perform supporting tasks such as monitoring, logging, or cleanup. When all user threads finish, daemon threads are automatically terminated.

Example

```python
import threading
import time

def background():
    while True:
        print("Daemon Thread Running")
        time.sleep(1)

t = threading.Thread(target=background)
t.daemon = True
t.start()

time.sleep(3)
print("Main Thread Finished")
```
