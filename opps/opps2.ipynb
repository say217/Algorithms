## Polymorphism
Definition

Polymorphism is an Object-Oriented Programming (OOP) concept that allows the same method or function name to perform different actions depending on the object that calls it. The word polymorphism comes from the Greek words "poly" (many) and "morph" (forms), meaning "many forms." It enables different classes to define their own implementation of the same method, making programs more flexible, reusable, and easier to maintain.

Simple Example


````python

class Dog:
    def sound(self):
        print("Dog barks.")


class Cat:
    def sound(self):
        print("Cat meows.")


animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
````

## Polymorphism – Method Overriding
Theory

Method overriding is a feature of polymorphism in which a child class provides its own implementation of a method that is already defined in the parent class. The method in the child class must have the same name and same parameters as the method in the parent class. When an object of the child class calls the method, the child class's version is executed instead of the parent class's version. Method overriding allows a program to perform different actions using the same method name, making the code more flexible, reusable, and easier to maintain.

Python Implementation


````python
# Parent class
class Animal:
    def sound(self):
        print("Animals make different sounds.")


# Child class
class Dog(Animal):
    def sound(self):
        print("Dog barks: Woof Woof!")


# Child class
class Cat(Animal):
    def sound(self):
        print("Cat meows: Meow Meow!")


# Create objects
d = Dog()
c = Cat()

# Call overridden methods
d.sound()
c.sound()
````

## Duck Typing
Theory

Duck typing is a feature of Python where the type of an object is determined by its behavior rather than its class. Instead of checking whether an object belongs to a particular class, Python checks whether the object has the required method or attribute. If an object implements the required method, it can be used, regardless of its class. This concept is based on the saying: "If it walks like a duck and quacks like a duck, then it is a duck." Duck typing makes Python flexible because different classes can be used interchangeably as long as they provide the expected behavior.

Python Implementation


```bash
Dog and Cat are different classes.
Both classes define a sound() method.
The make_sound() function does not check the object's class.
It simply calls animal.sound().
Since both objects have the required method, the function works for both.
This is called duck typing.

````python
class Dog:
    def sound(self):
        print("Dog barks.")


class Cat:
    def sound(self):
        print("Cat meows.")


def make_sound(animal):
    animal.sound()


d = Dog()
c = Cat()

make_sound(d)
make_sound(c)
````

## Operator Overloading
Theory

Operator overloading is a form of polymorphism in which the same operator performs different operations depending on the type of operands. Python allows programmers to redefine the behavior of operators for user-defined classes by implementing special methods such as __add__(), __sub__(), __mul__(), and others. This makes objects behave like built-in data types and allows operators to work in a meaningful way with custom classes.

Python Implementation


| Operator | Special Method             | Example             |
| -------- | -------------------------- | ------------------- |
| `+`      | `__add__(self, other)`     | Addition            |
| `-`      | `__sub__(self, other)`     | Subtraction         |
| `*`      | `__mul__(self, other)`     | Multiplication      |
| `/`      | `__truediv__(self, other)` | Division            |
| `==`     | `__eq__(self, other)`      | Equality comparison |
| `<`      | `__lt__(self, other)`      | Less than           |
| `>`      | `__gt__(self, other)`      | Greater than        |


````python

class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def display(self):
        print("Sum =", self.value)


n1 = Number(10)
n2 = Number(20)

result = n1 + n2

result.display()
````

## Method Overloading (Using Default Arguments or *args)
Theory

Method overloading is an OOP concept in which multiple methods have the same name but different numbers or types of parameters. In many programming languages such as Java and C++, method overloading is supported directly. However, Python does not support true method overloading because if multiple methods with the same name are defined in a class, the last definition replaces the previous ones.

To achieve similar functionality, Python uses default arguments, *args (variable-length arguments), or keyword arguments (**kwargs). These techniques allow a single method to accept different numbers of arguments and perform different operations based on the input. This provides flexibility while maintaining a single method name.

````python
class Calculator:
    def add(self, a, b=0):
        print("Sum =", a + b)


# Create object
c = Calculator()

# Call method with one argument
c.add(10)

# Call method with two arguments
c.add(10, 20)
````

```bash
Output
Sum = 10
Sum = 30
Explanation
The add() method has a default value for parameter b.
When only one argument is passed, b becomes 0.
When two arguments are passed, the provided value is used.
Thus, the same method behaves differently depending on the number of arguments.

````python
##!SECTIONMethod Overloading Using *args

class Calculator:
    def add(self, *args):
        print("Sum =", sum(args))


# Create object
c = Calculator()

c.add(10)
c.add(10, 20)
c.add(10, 20, 30)
c.add(10, 20, 30, 40)
````

```bash
Output
Sum = 10
Sum = 30
Sum = 60
Sum = 100
Explanation
*args allows a method to accept any number of positional arguments.
The arguments are stored as a tuple.
The built-in sum() function adds all the values.
The same add() method works with one, two, three, or more arguments

Why Python Does Not Support True Method Overloading

If multiple methods with the same name are defined, only the last method is kept.

## Why Python Does Not Support True Method Overloading

Python does not support true method overloading because method names in a class are stored as keys in the class namespace (a dictionary). Each key (method name) can have only one value (method implementation). If multiple methods with the same name are defined, the last method definition overwrites the previous ones. Unlike Java or C++, Python does not consider the number or types of parameters when identifying a method. Instead, Python resolves methods only by their name, so only one method with a given name can exist in a class. To achieve similar functionality, Python uses default arguments, *args, and **kwargs, allowing a single method to handle different numbers of arguments at runtime.

## Constructors

A constructor is a special method in a class that is automatically called when an object is created. Its main purpose is to initialize the object's data (attributes) and prepare the object for use. In Python, the constructor is defined using the __init__() method. Whenever an object is created, Python automatically executes the __init__() method. Constructors help assign initial values to variables, reduce repetitive code, and ensure every object starts in a valid state. A constructor can accept arguments, perform calculations, allocate resources, or initialize default values depending on the program's requirements.

## Default Constructor

A default constructor is a constructor that does not require any arguments from the user (except the mandatory self parameter). It initializes the object with predefined or default values. When an object is created, the constructor automatically assigns these default values to the instance variables. Default constructors are useful when every object should begin with the same initial state or when no external input is needed during object creation. They simplify object creation because the programmer only needs to create the object without passing any values.

````python
class Student:
    def __init__(self):
        self.name = "Unknown"
        self.age = 18

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student()
s1.display()
````

## Parameterized Constructor

A parameterized constructor is a constructor that accepts one or more arguments during object creation. These arguments are used to initialize the object's attributes with different values for different objects. This allows every object to store unique information instead of using the same default values. Parameterized constructors make programs more flexible because users can provide values when creating objects. They are commonly used in real-world applications where each object represents different data.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Rahul", 20)
s2 = Student("Anita", 21)

s1.display()
s2.display()
````

## Constructor Chaining

Constructor chaining is the process where one constructor calls another constructor to reuse initialization code instead of writing the same code multiple times. It improves code reusability, reduces duplication, and makes programs easier to maintain. In Python, constructor chaining is commonly achieved using the super() function, which calls the constructor of the parent (base) class from the child (derived) class. When an object of the child class is created, the parent constructor executes first to initialize inherited attributes, followed by the child constructor to initialize its own attributes.

- Working of Constructor Chaining
- An object of the child class is created.
- The child constructor (__init__()) starts executing.
- super().__init__() calls the parent class constructor.
- The parent constructor initializes the inherited attributes.
- Control returns to the child constructor.
- The child constructor initializes its own attributes.
- The object is fully initialized and ready to use.

Advantages of Constructor Chaining
- Avoids duplicate initialization code.
- Improves code reusability.
- Makes inheritance easier to manage.
- Simplifies maintenance and debugging.
- Ensures parent class attributes are initialized before child class attributes.

````python
class Person:
    def __init__(self, name):
        self.name = name
        print("Person constructor called")

class Student(Person):
    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll
        print("Student constructor called")

    def display(self):
        print("Name:", self.name)
        print("Roll:", self.roll)

s = Student("Rahul", 101)
s.display()
````

## Calling Parent Constructor using super() in Python

The super() function is used to call the constructor (__init__()) or methods of the parent (base) class from the child (derived) class. It allows the child class to inherit and initialize the parent class's attributes before initializing its own attributes. This avoids rewriting the same code, improves code reusability, and ensures proper initialization of inherited data.

```bash
Why Use super()?
Calls the parent constructor automatically.
Reuses parent initialization code.
Avoids writing duplicate code.
Makes inheritance easier to maintain.
Supports multiple inheritance correctly by following Python's Method Resolution Order (MRO).
Improves readability and maintainability.
Without super() (Not Recommended)
class Student(Person):
    def __init__(self, name, age, roll):
        Person.__init__(self, name, age)
        self.roll = roll

This works, but super() is preferred because it is more flexible, especially when working with multiple inheritance.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print("Person constructor called")

    def display_person(self):
        print("Name :", self.name)
        print("Age  :", self.age)


class Student(Person):
    def __init__(self, name, age, roll, course):
        # Calling parent constructor
        super().__init__(name, age)

        # Child class attributes
        self.roll = roll
        self.course = course
        print("Student constructor called")

    def display_student(self):
        print("\nStudent Details")
        print("----------------")
        print("Name   :", self.name)
        print("Age    :", self.age)
        print("Roll   :", self.roll)
        print("Course :", self.course)


# Creating object
s1 = Student("Rahul", 20, 101, "Computer Science")

print()

# Calling methods
s1.display_person()
print()
s1.display_student()
````

## Destructor

A destructor is a special method in Python that is automatically called when an object is about to be destroyed or removed from memory. Its main purpose is to perform cleanup operations before the object is deleted. These cleanup tasks may include closing files, releasing database connections, freeing network resources, or performing other necessary final operations. In Python, the destructor is defined using the __del__() method. Unlike constructors, which initialize an object when it is created, destructors execute when the object is no longer needed or when Python's garbage collector destroys the object.

## __del__() Method

The __del__() method is Python's built-in destructor. It is automatically invoked when an object's last reference is removed and the object is about to be destroyed. Programmers generally do not call __del__() directly. Instead, Python calls it automatically during object deletion. Although __del__() is useful for cleanup tasks, its execution time is not always predictable because it depends on Python's garbage collection mechanism. Therefore, for important resources such as files or database connections, context managers (with statement) are often preferred.

````python
class Student:
    def __init__(self, name):
        self.name = name
        print(f"Constructor called for {self.name}")

    def __del__(self):
        print(f"Destructor called for {self.name}")


# Creating object
s1 = Student("Rahul")

print("Program is running...")

# Deleting object
del s1

print("Object deleted")
````

## Object Cleanup

Object cleanup is the process of releasing or closing resources that an object was using before it is destroyed. During the lifetime of an object, it may hold resources such as open files, database connections, network sockets, or memory buffers. If these resources are not released properly, they may lead to memory leaks or resource wastage. The destructor (__del__()) is commonly used to perform these cleanup operations automatically when the object is destroyed.

````python
class FileManager:
    def __init__(self, filename):
        self.file = open(filename, "w")
        print("File opened")

    def write_data(self):
        self.file.write("Hello Python\n")

    def __del__(self):
        self.file.close()
        print("File closed during object cleanup")


# Create object
obj = FileManager("sample.txt")

obj.write_data()

# Delete object
del obj
````
