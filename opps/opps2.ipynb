## Polymorphism
Definition

Polymorphism is an Object-Oriented Programming (OOP) concept that allows the same method or function name to perform different actions depending on the object that calls it. The word polymorphism comes from the Greek words "poly" (many) and "morph" (forms), meaning "many forms." It enables different classes to define their own implementation of the same method, making programs more flexible, reusable, and easier to maintain.

Simple Example


````python

class Dog:
    def sound(self):
        print("Dog barks.")


class Cat:
    def sound(self):
        print("Cat meows.")


animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
````

## Polymorphism – Method Overriding
Theory

Method overriding is a feature of polymorphism in which a child class provides its own implementation of a method that is already defined in the parent class. The method in the child class must have the same name and same parameters as the method in the parent class. When an object of the child class calls the method, the child class's version is executed instead of the parent class's version. Method overriding allows a program to perform different actions using the same method name, making the code more flexible, reusable, and easier to maintain.

Python Implementation


````python
# Parent class
class Animal:
    def sound(self):
        print("Animals make different sounds.")


# Child class
class Dog(Animal):
    def sound(self):
        print("Dog barks: Woof Woof!")


# Child class
class Cat(Animal):
    def sound(self):
        print("Cat meows: Meow Meow!")


# Create objects
d = Dog()
c = Cat()

# Call overridden methods
d.sound()
c.sound()
````

## Duck Typing
Theory

Duck typing is a feature of Python where the type of an object is determined by its behavior rather than its class. Instead of checking whether an object belongs to a particular class, Python checks whether the object has the required method or attribute. If an object implements the required method, it can be used, regardless of its class. This concept is based on the saying: "If it walks like a duck and quacks like a duck, then it is a duck." Duck typing makes Python flexible because different classes can be used interchangeably as long as they provide the expected behavior.

Python Implementation


```bash
Dog and Cat are different classes.
Both classes define a sound() method.
The make_sound() function does not check the object's class.
It simply calls animal.sound().
Since both objects have the required method, the function works for both.
This is called duck typing.

````python
class Dog:
    def sound(self):
        print("Dog barks.")


class Cat:
    def sound(self):
        print("Cat meows.")


def make_sound(animal):
    animal.sound()


d = Dog()
c = Cat()

make_sound(d)
make_sound(c)
````

## Operator Overloading
Theory

Operator overloading is a form of polymorphism in which the same operator performs different operations depending on the type of operands. Python allows programmers to redefine the behavior of operators for user-defined classes by implementing special methods such as __add__(), __sub__(), __mul__(), and others. This makes objects behave like built-in data types and allows operators to work in a meaningful way with custom classes.

Python Implementation


| Operator | Special Method             | Example             |
| -------- | -------------------------- | ------------------- |
| `+`      | `__add__(self, other)`     | Addition            |
| `-`      | `__sub__(self, other)`     | Subtraction         |
| `*`      | `__mul__(self, other)`     | Multiplication      |
| `/`      | `__truediv__(self, other)` | Division            |
| `==`     | `__eq__(self, other)`      | Equality comparison |
| `<`      | `__lt__(self, other)`      | Less than           |
| `>`      | `__gt__(self, other)`      | Greater than        |


````python

class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def display(self):
        print("Sum =", self.value)


n1 = Number(10)
n2 = Number(20)

result = n1 + n2

result.display()
````

## Method Overloading (Using Default Arguments or *args)
Theory

Method overloading is an OOP concept in which multiple methods have the same name but different numbers or types of parameters. In many programming languages such as Java and C++, method overloading is supported directly. However, Python does not support true method overloading because if multiple methods with the same name are defined in a class, the last definition replaces the previous ones.

To achieve similar functionality, Python uses default arguments, *args (variable-length arguments), or keyword arguments (**kwargs). These techniques allow a single method to accept different numbers of arguments and perform different operations based on the input. This provides flexibility while maintaining a single method name.

````python
class Calculator:
    def add(self, a, b=0):
        print("Sum =", a + b)


# Create object
c = Calculator()

# Call method with one argument
c.add(10)

# Call method with two arguments
c.add(10, 20)
````

```bash
Output
Sum = 10
Sum = 30
Explanation
The add() method has a default value for parameter b.
When only one argument is passed, b becomes 0.
When two arguments are passed, the provided value is used.
Thus, the same method behaves differently depending on the number of arguments.

````python
##!SECTIONMethod Overloading Using *args

class Calculator:
    def add(self, *args):
        print("Sum =", sum(args))


# Create object
c = Calculator()

c.add(10)
c.add(10, 20)
c.add(10, 20, 30)
c.add(10, 20, 30, 40)
````

```bash
Output
Sum = 10
Sum = 30
Sum = 60
Sum = 100
Explanation
*args allows a method to accept any number of positional arguments.
The arguments are stored as a tuple.
The built-in sum() function adds all the values.
The same add() method works with one, two, three, or more arguments

Why Python Does Not Support True Method Overloading

If multiple methods with the same name are defined, only the last method is kept.

Example
```python
class Demo:

    def show(self):
        print("First method")

    def show(self, name):
        print("Hello", name)


d = Demo()

d.show("Alice")
```
```bash
Output
Hello Alice
```
If you try:

d.show()

You will get:

TypeError: Demo.show() missing 1 required positional argument: 'name'

Reason: The second show() method overrides the first one, so Python does not keep multiple methods with the same name.

## Why Python Does Not Support True Method Overloading

Python does not support true method overloading because method names in a class are stored as keys in the class namespace (a dictionary). Each key (method name) can have only one value (method implementation). If multiple methods with the same name are defined, the last method definition overwrites the previous ones. Unlike Java or C++, Python does not consider the number or types of parameters when identifying a method. Instead, Python resolves methods only by their name, so only one method with a given name can exist in a class. To achieve similar functionality, Python uses default arguments, *args, and **kwargs, allowing a single method to handle different numbers of arguments at runtime.
