## Instance vs Class

In object-oriented programming (OOP), a class is a blueprint or template used to create objects, while an instance is an actual object created from that class. The class defines the properties (variables) and behaviors (methods) that its objects will have. Each instance has its own copy of instance variables, allowing different objects of the same class to store different data. A class exists only once in a program, but many instances can be created from it. For example, if Student is a class, then Student("Rahul") and Student("Anita") are two different instances of that class.

````python
class Student:
    def __init__(self, name):
        self.name = name

# Creating instances (objects)
s1 = Student("Rahul")
s2 = Student("Anita")

print(s1.name)
print(s2.name)
````

## Instance Variables

Instance variables are variables that belong to an individual object (instance) of a class. They are created inside the constructor (__init__()) or other instance methods using the self keyword. Every object has its own separate copy of instance variables, so changing the value in one object does not affect other objects. Instance variables store data that is unique to each object.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Rahul", 20)
s2 = Student("Anita", 21)

print(s1.name, s1.age)
print(s2.name, s2.age)
````

## self.name

self.name is an instance variable. The keyword self refers to the current object on which a method is being called. When you write self.name = name, Python creates an instance variable named name and stores the value passed to the constructor in the current object. Every object has its own self.name, allowing each object to maintain different data independently.

| Feature  | Class                         | Instance                                     |
| -------- | ----------------------------- | -------------------------------------------- |
| Meaning  | Blueprint or template         | Actual object created from the class         |
| Creation | Defined using `class` keyword | Created using the class name                 |
| Number   | One class can exist           | Many instances can be created                |
| Data     | Defines structure and methods | Stores its own data using instance variables |


````python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student Name:", self.name)

s1 = Student("Rahul")
s2 = Student("Anita")

s1.display()
s2.display()
````

## Class Variables

A class variable is a variable that belongs to the class itself, not to individual objects (instances). It is declared inside the class but outside all methods. A single copy of the class variable is shared by all objects created from that class. If the value of a class variable is changed, the updated value is reflected in all objects (unless an object creates its own instance variable with the same name). Class variables are commonly used to store data that is common to every object, such as a school name, company name, or tax rate.

## Types of Methods

In Python, methods are functions defined inside a class that describe the behavior of objects. Depending on how they access data, methods are classified into Instance Methods, Class Methods, and Static Methods. Each type serves a different purpose. Instance methods work with object-specific data, class methods work with class-level data, and static methods perform utility operations that do not depend on either instance or class data.

## Instance Method

An instance method is the most common type of method in Python. It operates on a specific object (instance) of a class and can access or modify that object's instance variables. An instance method always takes self as its first parameter, where self refers to the current object that calls the method. These methods are used when the operation depends on the data stored in an individual object, such as displaying details, updating attributes, or performing calculations using instance variables.

## Class Method

A class method is a method that operates on the class itself rather than on individual objects. It is defined using the @classmethod decorator and takes cls as its first parameter, where cls refers to the class. Class methods can access and modify class variables but cannot directly access instance variables unless an object is provided. They are commonly used for operations related to the entire class, such as creating alternative constructors, modifying shared class data, or performing actions that affect all objects of the class.

## Static Method

A static method is a method that belongs to a class but does not depend on either the object (self) or the class (cls). It is defined using the @staticmethod decorator and does not take self or cls as its first parameter. Static methods behave like normal functions placed inside a class for better organization. They cannot directly access instance variables or class variables unless they are explicitly passed as arguments. Static methods are typically used for utility or helper functions, such as mathematical calculations, data validation, or conversions, that are logically related to the class but do not require access to its data.

````python
# instance  
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance Method
    def display(self):
        print("Name:", self.name)
        print("Age :", self.age)

# Create object
s1 = Student("Rahul", 20)

# Call instance method
s1.display()
````

````python
# class method 
class Student:
    school = "ABC School"

    @classmethod
    def display_school(cls):
        print("School Name:", cls.school)

# Call class method
Student.display_school()
````

````python
class Student:
    school = "ABC School"

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

Student.change_school("XYZ School")

print(Student.school)
````

````python
# 3. Static Method Example


class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

result = Calculator.add(10, 20)

print("Sum =", result)
````

````python
class Student:

    # Class Variable
    school = "ABC School"

    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance Method
    def display(self):
        print("Name  :", self.name)
        print("Age   :", self.age)
        print("School:", Student.school)

    # Class Method
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

    # Static Method
    @staticmethod
    def greet():
        print("Welcome to the Student Management System")


# Create object
s1 = Student("Rahul", 20)

# Static Method
Student.greet()

print()

# Instance Method
s1.display()

print()

# Class Method
Student.change_school("XYZ School")

# Check updated class variable
s1.display()
````

## Name Mangling

Name mangling is a mechanism in Python that changes the name of a class attribute or method when it begins with two leading underscores (__). Python automatically renames the variable internally to prevent it from being accidentally accessed or overridden by subclasses. Name mangling is not true data hiding or security; it is mainly used to avoid naming conflicts in inheritance and to indicate that a variable or method is intended for internal use within the class.

The internal name is changed using the format:

````python
# Name Mangling
class Student:
    def __init__(self):
        self.__name = "Rahul"

    def display(self):
        print("Name:", self.__name)

s = Student()

# Access through method
s.display()

# Direct access (will cause an error)
# print(s.__name)
````

````python
# Accessing the Mangled Name
class Student:
    def __init__(self):
        self.__name = "Rahul"

s = Student()

print(s._Student__name)
````

````python
# Preventing Name Conflicts in Inheritance
class Person:
    def __init__(self):
        self.__name = "Person"

    def show(self):
        print(self.__name)


class Student(Person):
    def __init__(self):
        super().__init__()
        self.__name = "Student"

    def display(self):
        print(self.__name)


s = Student()

s.show()
s.display()
````

## Special methods

````python
class Student:

    # Constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # String representation (for users)
    def __str__(self):
        return f"Student(Name={self.name}, Marks={self.marks})"

    # Official representation (for developers)
    def __repr__(self):
        return f"Student('{self.name}', {self.marks})"

    # Length of object
    def __len__(self):
        return len(self.name)

    # Addition operator
    def __add__(self, other):
        return self.marks + other.marks


# Create objects
s1 = Student("Rahul", 85)
s2 = Student("Anita", 90)

# __str__()
print(s1)

# __repr__()
print(repr(s1))

# __len__()
print("Length of name:", len(s1))

# __add__()
print("Total Marks:", s1 + s2)
````

````python
class Number:

    def __init__(self, value):
        self.value = value

    # Subtraction (-)
    def __sub__(self, other):
        return self.value - other.value

    # Multiplication (*)
    def __mul__(self, other):
        return self.value * other.value

    # Equal to (==)
    def __eq__(self, other):
        return self.value == other.value

    # Less than (<)
    def __lt__(self, other):
        return self.value < other.value

    # Greater than (>)
    def __gt__(self, other):
        return self.value > other.value


n1 = Number(20)
n2 = Number(10)
n3 = Number(20)

print("Subtraction :", n1 - n2)
print("Multiplication :", n1 * n2)
print("Equal :", n1 == n3)
print("Less Than :", n1 < n2)
print("Greater Than :", n1 > n2)
````

| Magic Method     | Automatically Called By         | Purpose                                       |
| ---------------- | ------------------------------- | --------------------------------------------- |
| `__getitem__()`  | `obj[index]`                    | Returns an item using `[]`                    |
| `__setitem__()`  | `obj[index] = value`            | Updates an item using `[]`                    |
| `__contains__()` | `value in obj`                  | Checks if an item exists                      |
| `__call__()`     | `obj()`                         | Makes an object callable like a function      |
| `__iter__()`     | `iter(obj)` or `for` loop       | Returns an iterator                           |
| `__next__()`     | `next(obj)`                     | Returns the next item during iteration        |
| `__del__()`      | `del obj` or garbage collection | Cleans up resources before object destruction |


````python
class Demo:

    def __init__(self):
        self.data = [10, 20, 30]
        self.index = 0

    # Access item using []
    def __getitem__(self, index):
        return self.data[index]

    # Set item using []
    def __setitem__(self, index, value):
        self.data[index] = value

    # Check membership using in
    def __contains__(self, value):
        return value in self.data

    # Make object callable
    def __call__(self):
        print("Object called like a function")

    # Return iterator
    def __iter__(self):
        self.index = 0
        return self

    # Return next element
    def __next__(self):
        if self.index < len(self.data):
            value = self.data[self.index]
            self.index += 1
            return value
        raise StopIteration

    # Destructor
    def __del__(self):
        print("Object destroyed")


obj = Demo()

# __getitem__()
print("Item:", obj[1])

# __setitem__()
obj[1] = 100
print("Updated List:", obj.data)

# __contains__()
print(100 in obj)

# __call__()
obj()

# __iter__() and __next__()
print("Iteration:")
for i in obj:
    print(i)

# __del__()
del obj
````

## super()

The super() function is a built-in Python function used to access the methods and constructor of the parent (base) class from the child (derived) class. It allows the child class to reuse the functionality already defined in the parent class instead of rewriting the same code. The use of super() supports code reusability, simplifies inheritance, and ensures that the parent class is properly initialized before the child class adds its own functionality. It is especially useful in inheritance because it follows Python's Method Resolution Order (MRO), making it reliable even in multiple inheritance.

## Calling Parent Constructor

A parent constructor is the __init__() method of the parent class. When a child class defines its own constructor, the parent constructor is not called automatically. To initialize the parent class attributes, the child class must explicitly call the parent constructor using super().__init__(). This ensures that all instance variables defined in the parent class are properly initialized before the child class initializes its own variables. Calling the parent constructor avoids duplicate code and allows the child class to inherit the parent's initialization logic.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print("Person constructor called")


class Student(Person):
    def __init__(self, name, age, roll):
        super().__init__(name, age)   # Calling parent constructor
        self.roll = roll
        print("Student constructor called")


s = Student("Rahul", 20, 101)

print(s.name)
print(s.age)
print(s.roll)
````

## Calling Parent Methods

A parent method is any method defined in the parent class other than the constructor. A child class can call these methods using super().method_name(). This allows the child class to reuse the parent's implementation and, if needed, add extra functionality before or after the parent method is executed. Calling parent methods is useful when the child class wants to extend the parent's behavior instead of replacing it completely.

```bash
Advantages of super()
Reuses code from the parent class.
Avoids duplicate code.
Ensures the parent constructor is properly executed.
Makes inheritance easier to maintain.
Supports multiple inheritance correctly through Python's Method Resolution Order (MRO).
Improves code readability and maintainability.

````python
class Person:
    def show(self):
        print("This is the Person class.")


class Student(Person):
    def show(self):
        super().show()      # Calling parent method
        print("This is the Student class.")


s = Student()
s.show()
````

## Method Resolution Order (MRO)

Method Resolution Order (MRO) is the order in which Python searches for methods and attributes when they are called on an object, especially in inheritance. MRO becomes particularly important in multiple inheritance, where a child class inherits from more than one parent class. Python follows a specific sequence to determine which method should be executed if the same method exists in multiple parent classes. Python uses the C3 Linearization Algorithm to calculate the MRO, ensuring a consistent and predictable order while avoiding ambiguity.

## Diamond Problem

The Diamond Problem is a situation in multiple inheritance where a class inherits from two parent classes, and both of those parent classes inherit from the same grandparent class. This creates a diamond-shaped inheritance structure. If the grandparent class contains a method that is overridden in both parent classes, Python must decide which version of the method to execute. Without a proper resolution strategy, this could cause ambiguity or duplicate method calls. Python solves this problem using Method Resolution Order (MRO) and the super() function, ensuring that each class is visited only once and in a well-defined order.

Example

````python
class A:
    def show(self):
        print("Class A")


class B(A):
    def show(self):
        print("Class B")


class C(A):
    def show(self):
        print("Class C")


class D(B, C):
    pass


obj = D()
obj.show()
````

## Class.mro()

The mro() method is a built-in class method that returns the Method Resolution Order as a list of classes. It helps programmers understand the exact order in which Python searches for methods and attributes during inheritance. It is mainly used for debugging, learning inheritance behavior, and verifying the order followed by Python in complex inheritance hierarchies.

````python
class A:
    pass


class B(A):
    pass


class C(A):
    pass


class D(B, C):
    pass


print(D.mro())
````

````python
class A:
    def show(self):
        print("Class A")


class B(A):
    def show(self):
        print("Class B")
        super().show()


class C(A):
    def show(self):
        print("Class C")
        super().show()


class D(B, C):
    def show(self):
        print("Class D")
        super().show()


obj = D()
obj.show()
````
