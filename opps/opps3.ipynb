## Instance vs Class

In object-oriented programming (OOP), a class is a blueprint or template used to create objects, while an instance is an actual object created from that class. The class defines the properties (variables) and behaviors (methods) that its objects will have. Each instance has its own copy of instance variables, allowing different objects of the same class to store different data. A class exists only once in a program, but many instances can be created from it. For example, if Student is a class, then Student("Rahul") and Student("Anita") are two different instances of that class.

````python
class Student:
    def __init__(self, name):
        self.name = name

# Creating instances (objects)
s1 = Student("Rahul")
s2 = Student("Anita")

print(s1.name)
print(s2.name)
````

## Instance Variables

Instance variables are variables that belong to an individual object (instance) of a class. They are created inside the constructor (__init__()) or other instance methods using the self keyword. Every object has its own separate copy of instance variables, so changing the value in one object does not affect other objects. Instance variables store data that is unique to each object.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Rahul", 20)
s2 = Student("Anita", 21)

print(s1.name, s1.age)
print(s2.name, s2.age)
````

## self.name

self.name is an instance variable. The keyword self refers to the current object on which a method is being called. When you write self.name = name, Python creates an instance variable named name and stores the value passed to the constructor in the current object. Every object has its own self.name, allowing each object to maintain different data independently.

| Feature  | Class                         | Instance                                     |
| -------- | ----------------------------- | -------------------------------------------- |
| Meaning  | Blueprint or template         | Actual object created from the class         |
| Creation | Defined using `class` keyword | Created using the class name                 |
| Number   | One class can exist           | Many instances can be created                |
| Data     | Defines structure and methods | Stores its own data using instance variables |


````python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student Name:", self.name)

s1 = Student("Rahul")
s2 = Student("Anita")

s1.display()
s2.display()
````

## Class Variables

A class variable is a variable that belongs to the class itself, not to individual objects (instances). It is declared inside the class but outside all methods. A single copy of the class variable is shared by all objects created from that class. If the value of a class variable is changed, the updated value is reflected in all objects (unless an object creates its own instance variable with the same name). Class variables are commonly used to store data that is common to every object, such as a school name, company name, or tax rate.

```python
Syntax
class ClassName:
    class_variable = value

| Class Variable                             | Instance Variable                    |
| ------------------------------------------ | ------------------------------------ |
| Belongs to the class                       | Belongs to an object                 |
| Shared by all objects                      | Separate copy for each object        |
| Declared inside the class, outside methods | Declared using `self` inside methods |
| Example: `school = "ABC School"`           | Example: `self.name = "Rahul"`       |


````python
class Student:
    school = "ABC School"      # Class Variable

    def __init__(self, name):
        self.name = name       # Instance Variable

s1 = Student("Rahul")
s2 = Student("Anita")

print(s1.name, "-", s1.school)
print(s2.name, "-", s2.school)
````

## Types of Methods

In Python, methods are functions defined inside a class that describe the behavior of objects. Depending on how they access data, methods are classified into Instance Methods, Class Methods, and Static Methods. Each type serves a different purpose. Instance methods work with object-specific data, class methods work with class-level data, and static methods perform utility operations that do not depend on either instance or class data.

## Instance Method

An instance method is the most common type of method in Python. It operates on a specific object (instance) of a class and can access or modify that object's instance variables. An instance method always takes self as its first parameter, where self refers to the current object that calls the method. These methods are used when the operation depends on the data stored in an individual object, such as displaying details, updating attributes, or performing calculations using instance variables.

## Class Method

A class method is a method that operates on the class itself rather than on individual objects. It is defined using the @classmethod decorator and takes cls as its first parameter, where cls refers to the class. Class methods can access and modify class variables but cannot directly access instance variables unless an object is provided. They are commonly used for operations related to the entire class, such as creating alternative constructors, modifying shared class data, or performing actions that affect all objects of the class.

## Static Method

A static method is a method that belongs to a class but does not depend on either the object (self) or the class (cls). It is defined using the @staticmethod decorator and does not take self or cls as its first parameter. Static methods behave like normal functions placed inside a class for better organization. They cannot directly access instance variables or class variables unless they are explicitly passed as arguments. Static methods are typically used for utility or helper functions, such as mathematical calculations, data validation, or conversions, that are logically related to the class but do not require access to its data.

````python
# instance  
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance Method
    def display(self):
        print("Name:", self.name)
        print("Age :", self.age)

# Create object
s1 = Student("Rahul", 20)

# Call instance method
s1.display()
````

````python
# class method 
class Student:
    school = "ABC School"

    @classmethod
    def display_school(cls):
        print("School Name:", cls.school)

# Call class method
Student.display_school()
````

````python
class Student:
    school = "ABC School"

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

Student.change_school("XYZ School")

print(Student.school)
````

````python
# 3. Static Method Example


class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

result = Calculator.add(10, 20)

print("Sum =", result)
````

````python
class Student:

    # Class Variable
    school = "ABC School"

    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance Method
    def display(self):
        print("Name  :", self.name)
        print("Age   :", self.age)
        print("School:", Student.school)

    # Class Method
    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

    # Static Method
    @staticmethod
    def greet():
        print("Welcome to the Student Management System")


# Create object
s1 = Student("Rahul", 20)

# Static Method
Student.greet()

print()

# Instance Method
s1.display()

print()

# Class Method
Student.change_school("XYZ School")

# Check updated class variable
s1.display()
````

## Name Mangling

Name mangling is a mechanism in Python that changes the name of a class attribute or method when it begins with two leading underscores (__). Python automatically renames the variable internally to prevent it from being accidentally accessed or overridden by subclasses. Name mangling is not true data hiding or security; it is mainly used to avoid naming conflicts in inheritance and to indicate that a variable or method is intended for internal use within the class.

The internal name is changed using the format:

````python
# Name Mangling
class Student:
    def __init__(self):
        self.__name = "Rahul"

    def display(self):
        print("Name:", self.__name)

s = Student()

# Access through method
s.display()

# Direct access (will cause an error)
# print(s.__name)
````

````python
# Accessing the Mangled Name
class Student:
    def __init__(self):
        self.__name = "Rahul"

s = Student()

print(s._Student__name)
````

````python
# Preventing Name Conflicts in Inheritance
class Person:
    def __init__(self):
        self.__name = "Person"

    def show(self):
        print(self.__name)


class Student(Person):
    def __init__(self):
        super().__init__()
        self.__name = "Student"

    def display(self):
        print(self.__name)


s = Student()

s.show()
s.display()
````

## Special methods

````python
class Student:

    # Constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # String representation (for users)
    def __str__(self):
        return f"Student(Name={self.name}, Marks={self.marks})"

    # Official representation (for developers)
    def __repr__(self):
        return f"Student('{self.name}', {self.marks})"

    # Length of object
    def __len__(self):
        return len(self.name)

    # Addition operator
    def __add__(self, other):
        return self.marks + other.marks


# Create objects
s1 = Student("Rahul", 85)
s2 = Student("Anita", 90)

# __str__()
print(s1)

# __repr__()
print(repr(s1))

# __len__()
print("Length of name:", len(s1))

# __add__()
print("Total Marks:", s1 + s2)
````
